```python
# -----######-----###### CONCAT 3 PKL FILES INTO ONE DF -----######-----###### #
import pandas as pd

def _concat_1710_pkls_GET_combined_df(pkl_paths):
    """
    Concatenate 3 DataFrames from provided .pkl file paths into one DataFrame.
    Parameters:
        pkl_paths: list of 3 strings (file paths to .pkl files)
    Returns:
        pd.DataFrame: concatenated DataFrame
    """
    dfs = []
    for path in pkl_paths:
        print(f"📥 Reading: {path}")
        df = pd.read_pickle(path)
        print(f"✅ Loaded {len(df)} rows")
        dfs.append(df)

    df_final = pd.concat(dfs, ignore_index=True)
    print(f"\n🎯 Total Rows Combined: {len(df_final)}")
    return df_final
```

```python
# -----######-----###### MP3 PATHS + BASENAME 2-LEVEL -----######-----###### #
import os
import pandas as pd
from tqdm import tqdm

def _paths_1707_2levelmp3_GET_df(folder_root):
    """
    Traverses 2 levels deep and collects all .mp3 files.
    Returns a DataFrame with:
        - 'Path': full path to the file
        - 'file_name': filename with extension
        - 'Name': base name without extension
    """
    paths = []
    file_names = []
    base_names = []

    print(f"📂 Searching inside: {folder_root}")

    for lvl1 in tqdm(sorted(os.listdir(folder_root))):
        lvl1_path = os.path.join(folder_root, lvl1)
        if not os.path.isdir(lvl1_path): continue

        for lvl2 in sorted(os.listdir(lvl1_path)):
            lvl2_path = os.path.join(lvl1_path, lvl2)
            if not os.path.isdir(lvl2_path): continue

            for file in os.listdir(lvl2_path):
                if file.lower().endswith(".mp3") and not file.startswith("._"):
                    full_path = os.path.join(lvl2_path, file)
                    paths.append(full_path)
                    file_names.append(file)
                    base_names.append(os.path.splitext(file)[0])

    df = pd.DataFrame({
        "Path": paths,
        "file_name": file_names,
        "base_name": base_names
    })

    print(f"🎯 Total MP3s found: {len(df)}")
    return df
folder_input = "/Users/yerik/Music/_4_MUSIC_PROD/_0_PROD_MATERIAL/_ALL_3_PROD_stems_S"
df_mp3s = _paths_1707_2levelmp3_GET_df(folder_input)
```

```python
# -----######-----###### EXTRACT base_name FROM Name -----######-----###### #
def _parse_1707_name_GET_base_name(df):
    """
    From df['Name'], extract 'base_name' by:
    - Keeping first 3 characters
    - Appending characters up to the next '-' after that
    Adds column 'base_name'
    """
    base_names = []

    for name in df["file_name"]:
        prefix = name[:3]
        rest = name[3:]
        mid = rest.split("-")[0]
        base_name = f"{prefix}{mid}"
        base_names.append(base_name)

    df["ID"] = base_names
    print(f"🔡 base_name extracted for {len(df)} rows.")
    return df
```

```python
# -----######-----###### COMPARE TWO ID COLUMNS -----######-----###### #
def _compare_1707_ids_GET_mismatch_report(df1, df2, col_name="ID"):
    """
    Compares col_name in df1 and df2:
    - Checks if both sets are identical
    - Lists missing values from each side
    - Reports count summary
    """
    set1 = set(df1[col_name].dropna())
    set2 = set(df2[col_name].dropna())

    only_in_df1 = sorted(set1 - set2)
    only_in_df2 = sorted(set2 - set1)
    common_ids = sorted(set1 & set2)

    print(f"🔍 Comparing column: '{col_name}'")
    print(f"✅ Common IDs: {len(common_ids)}")
    print(f"❌ Only in df_mp3s: {len(only_in_df1)}")
    print(f"❌ Only in df_combined: {len(only_in_df2)}")

    if only_in_df1:
        print("📁 Missing in df_combined:")
        for val in only_in_df1[:10]:
            print(f"   - {val}")
        if len(only_in_df1) > 10:
            print("   ...")

    if only_in_df2:
        print("📁 Missing in df_mp3s:")
        for val in only_in_df2[:10]:
            print(f"   - {val}")
        if len(only_in_df2) > 10:
            print("   ...")

    return {
        "only_in_df_mp3s": only_in_df1,
        "only_in_df_combined": only_in_df2,
        "common_ids": common_ids
    }
```

```python
# -----######-----###### REPLACE PATHS FROM MATCHING ID -----######-----###### #
def _merge_1707_fixpaths_GET_dfnew(df_combined, df_mp3s):
    """
    Replaces df_combined['Path'] with the correct value from df_mp3s based on 'ID' match.
    Returns a new DataFrame identical to df_combined, but with updated 'Path'.
    """
    # Keep original just in case
    df_combined_updated = df_combined.copy()

    # Build dict: ID -> Path from df_mp3s
    id_to_path = dict(zip(df_mp3s["ID"], df_mp3s["Path"]))

    # Replace paths in df_combined
    df_combined_updated["Path"] = df_combined_updated["ID"].map(id_to_path)

    print(f"✅ Replaced Path values using ID match. Total rows updated: {len(df_combined_updated)}")
    return df_combined_updated
```

```python
df_combined_fixed = _merge_1707_fixpaths_GET_dfnew(df_combined, df_mp3s)
```

```python
# -----######-----###### VERIFY FILE PATHS EXIST -----######-----###### #
import os

def _check_1707_paths_GET_missing_summary(df, path_col="Path"):
    """
    Verifies which paths exist on disk for the given path column.
    Adds 'exists' column (True/False) and returns summary + missing paths.
    """
    print(f"🔎 Verifying {len(df)} paths in column: '{path_col}'")

    df["exists"] = df[path_col].apply(os.path.exists)
    total_missing = (~df["exists"]).sum()
    
    print(f"✅ Valid paths: {len(df) - total_missing}")
    print(f"❌ Missing paths: {total_missing}")

    if total_missing > 0:
        print("⚠️ Sample missing paths:")
        print(df[~df["exists"]][path_col].head(5).to_string(index=False))

    return df
```
