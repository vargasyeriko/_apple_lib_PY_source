```python
# -----######-----###### MP3 COUNT 2-LEVEL ONLY -----######-----###### #
import os

def _count_1707_mp3_2lvls_GET_total(folder_root):
    count = 0
    for lvl1 in os.listdir(folder_root):
        p1 = os.path.join(folder_root, lvl1)
        if not os.path.isdir(p1): continue
        for lvl2 in os.listdir(p1):
            p2 = os.path.join(p1, lvl2)
            if not os.path.isdir(p2): continue
            count += sum(1 for f in os.listdir(p2) if f.lower().endswith(".mp3"))
    print(f"🎯 MP3 files (2-levels only): {count}")
    return count
```

```python
# -----######-----###### MP3 COUNT PER FOLDER (4-LEVEL) -----######-----###### #
import os
import pandas as pd

def _count_1707_mp3_4lvls_GET_df(folder_root):
    """
    Explore folder_root > sub1 > sub2 > sub3 > files
    Count how many .mp3 files exist in each 4th-level folder
    """
    records = []
    for sub1 in os.listdir(folder_root):
        p1 = os.path.join(folder_root, sub1)
        if not os.path.isdir(p1): continue
        for sub2 in os.listdir(p1):
            p2 = os.path.join(p1, sub2)
            if not os.path.isdir(p2): continue
            for sub3 in os.listdir(p2):
                p3 = os.path.join(p2, sub3)
                if not os.path.isdir(p3): continue
                mp3s = [f for f in os.listdir(p3) if f.lower().endswith(".mp3")]
                records.append({
                    'Folder': p3,
                    'MP3_Count': len(mp3s),
                    'Is_Complete': len(mp3s) == 22
                })
    df = pd.DataFrame(records)
    print(f"🎯 Total 4th-level folders scanned: {len(df)}")
    print(f"✅ Folders with exactly 22 MP3s: {(df['Is_Complete']).sum()}")
    print(f"❌ Incomplete folders: {(~df['Is_Complete']).sum()}")
    return df
```

```python
folder_path = "/Users/yerik/Music/_4_MUSIC_PROD/_0_PROD_MATERIAL/_ALL_3_PROD_stems_S"
df_mp3_folders = _count_1707_mp3_4lvls_GET_df(folder_path)
```

```python
# Show only incomplete folders (not 22 mp3s)
df_incomplete = df_mp3_folders[df_mp3_folders['Is_Complete'] == False]
print("❌ Incomplete Folders (≠ 22 MP3s):")
print(df_incomplete[['Folder', 'MP3_Count']])
```
