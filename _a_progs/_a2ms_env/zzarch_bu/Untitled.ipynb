```python
# -----######-----###### GET AIFF COMMENTS FROM FOLDER -----######-----###### #
import os
from mutagen.aiff import AIFF
from pathlib import Path
from tqdm import tqdm

def _aiff_0808_comments_GET_df_comments(folder_path):
    folder = Path(folder_path)
    if not folder.exists():
        print("❌ Folder not found:", folder)
        return

    aiff_files = list(folder.rglob("*.aiff"))
    results = []

    for path in tqdm(aiff_files, desc="🔍 Scanning AIFF files"):
        try:
            audio = AIFF(path)
            comment = audio.get("COMM", [""])[0]
        except Exception as e:
            comment = f"⚠️ Error: {str(e)}"
        
        results.append({
            "Path": str(path),
            "File Name": path.name,
            "Comment": comment
        })

    import pandas as pd
    df = pd.DataFrame(results)
    return df
```

```python

# Replace with your actual folder containing AIFFs
folder_path = "/Users/yerik/Music/_1_NEW_SOURCE/_2025_this/__25_05_ULatin_MVMNT"
df_comments = _aiff_0808_comments_GET_df_comments(folder_path)

# View results
print(df_comments.head(10))
```
