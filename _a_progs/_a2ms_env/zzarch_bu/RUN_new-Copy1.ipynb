```python
# -----######-----######-----######-----######-----######-----#
import os
from tqdm import tqdm

def _dir_1704_f1_GET_list_subfolders(folder_path):
    """
    Walks through the given folder and returns a list of all subfolder paths.
    
    Parameters:
        folder_path: str - the root folder to scan
    
    Returns:
        List of full paths to all subfolders (recursively)
    """
    subfolders = []
    for root, dirs, _ in tqdm(os.walk(folder_path), desc="Scanning folders"):
        for d in dirs:
            subfolders.append(os.path.join(root, d))
    return subfolders

folder_path = '/Volumes/MUSIC_PROD/_1_NEW_SOURCE/_2025_this' # update this
subfolder_paths = _dir_1704_f1_GET_list_subfolders(folder_path)
subfolder_paths 
```
