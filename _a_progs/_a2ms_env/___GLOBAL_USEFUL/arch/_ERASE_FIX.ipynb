```python
# -----######-----###### COPY FILES: READ LIKE RENAMER FUNCTION -----######-----###### #
import os
import shutil
import pandas as pd
from tqdm import tqdm

def _fileops_2207_kwstyle_GET_copy_from_txt(txt_path, target_dir):
    """
    Reads a UTF-16 TXT file (tab-separated), looks for 'Path' or 'Location',
    and copies those files to the target_dir.
    """
    tqdm.pandas()

    # 🚀 Load with same forgiving style
    df = pd.read_csv(
        txt_path,
        sep="\t",
        encoding='utf-16',
        engine='python',
        on_bad_lines='skip'
    )
    df.columns = df.columns.str.strip()

    # ✅ Try to find either 'Path' or 'Location' column
    col_name = None
    for option in ['Path', 'Location']:
        if option in df.columns:
            col_name = option
            break

    if not col_name:
        print("❌ No 'Path' or 'Location' column found.")
        return pd.DataFrame()

    # 📂 Ensure target dir exists
    os.makedirs(target_dir, exist_ok=True)

    copied_paths = []
    for i, row in tqdm(df.iterrows(), total=len(df), desc="📁 Copying files"):
        src = str(row.get(col_name, '')).strip()
        if not os.path.isfile(src):
            copied_paths.append(None)
            continue
        try:
            dst = os.path.join(target_dir, os.path.basename(src))
            shutil.copy2(src, dst)
            copied_paths.append(dst)
        except Exception as e:
            print(f"❌ Row {i}: Failed to copy {src} → {e}")
            copied_paths.append(None)

    df['Copied_Path'] = copied_paths
    return df
```

```python
# -----######-----###### MOVE FILES: READ LIKE RENAMER FUNCTION -----######-----###### #
import os
import shutil
import pandas as pd
from tqdm import tqdm

def _fileops_2207_kwstyle_GET_move_from_txt(txt_path, target_dir):
    """
    Reads a UTF-16 TXT file (tab-separated), looks for 'Path' or 'Location',
    and moves those files to the target_dir.
    """
    tqdm.pandas()

    # 🚀 Load with same forgiving style
    df = pd.read_csv(
        txt_path,
        sep="\t",
        encoding='utf-16',
        engine='python',
        on_bad_lines='skip'
    )
    df.columns = df.columns.str.strip()

    # ✅ Try to find either 'Path' or 'Location' column
    col_name = None
    for option in ['Path', 'Location']:
        if option in df.columns:
            col_name = option
            break

    if not col_name:
        print("❌ No 'Path' or 'Location' column found.")
        return pd.DataFrame()

    # 📂 Ensure target dir exists
    os.makedirs(target_dir, exist_ok=True)

    moved_paths = []
    for i, row in tqdm(df.iterrows(), total=len(df), desc="📁 Moving files"):
        src = str(row.get(col_name, '')).strip()
        if not os.path.isfile(src):
            moved_paths.append(None)
            continue
        try:
            dst = os.path.join(target_dir, os.path.basename(src))
            shutil.move(src, dst)
            moved_paths.append(dst)
        except Exception as e:
            print(f"❌ Row {i}: Failed to move {src} → {e}")
            moved_paths.append(None)

    df['Moved_Path'] = moved_paths
    return df
```

```python
# -----######-----###### CHECK IF ERASE.TXT FILES EXIST IN IN_COLL.TXT -----######-----###### #
import pandas as pd

def _compare_2207_txts_GET_overlap_flags(txt_erase, txt_in_coll):
    """
    Compare two UTF-16 tab-separated TXT files and check which paths in erase.txt
    are also in in_coll.txt.

    Returns a DataFrame with:
        - Path_erase
        - Found_in_in_coll (True/False)
    """
    # Read both files (robustly)
    df_erase = pd.read_csv(txt_erase, sep="\t", encoding="utf-16", engine="python", on_bad_lines='skip')
    df_in = pd.read_csv(txt_in_coll, sep="\t", encoding="utf-16", engine="python", on_bad_lines='skip')

    # Clean column names
    df_erase.columns = df_erase.columns.str.strip()
    df_in.columns = df_in.columns.str.strip()

    # Choose column: Path or Location
    col_erase = next((c for c in ['Path', 'Location'] if c in df_erase.columns), df_erase.columns[0])
    col_in = next((c for c in ['Path', 'Location'] if c in df_in.columns), df_in.columns[0])

    # Create sets for fast lookup
    in_coll_paths = set(df_in[col_in].dropna().astype(str).str.strip())

    # Flag which erase paths are in in_coll
    df_erase['Path_erase'] = df_erase[col_erase].astype(str).str.strip()
    df_erase['Found_in_in_coll'] = df_erase['Path_erase'].isin(in_coll_paths)

    return df_erase[['Path_erase', 'Found_in_in_coll']]
```

```python
# -----######-----###### FILE COUNT PER FIRST-LEVEL FOLDER -----######-----###### #
import os
import pandas as pd
from tqdm import tqdm

def _paths_2207_i1_GET_count_files_firstlevel(folder_root):
    """
    Goes inside every first-level folder under folder_root
    and counts the number of files (non-directory) inside each.
    Returns a DataFrame with folder name, full path, and count.
    """
    data = []

    # Get only first-level directories
    first_level_folders = [f for f in os.listdir(folder_root) if os.path.isdir(os.path.join(folder_root, f))]

    for folder in tqdm(first_level_folders, desc="Counting files"):
        full_path = os.path.join(folder_root, folder)
        file_count = sum(
            1 for item in os.listdir(full_path)
            if os.path.isfile(os.path.join(full_path, item))
        )
        data.append({
            "Folder": folder,
            "Full_Path": full_path,
            "File_Count": file_count
        })

    return pd.DataFrame(data)
folder_root = "/Users/yerik/Music/_0_OLD_SOURCE"
df_counts = _paths_2207_i1_GET_count_files_firstlevel(folder_root)
df_counts.sort_values("File_Count", ascending=False)
```

```python
# -----######-----###### DELETE EMPTY FIRST-LEVEL FOLDERS -----######-----###### #
import os
import pandas as pd
from tqdm import tqdm

def _paths_2207_i2_GET_delete_empty_firstlevel(folder_root):
    """
    Checks all first-level folders in `folder_root` for zero files (not subfolders).
    Displays list and count of empty ones, then asks for confirmation to delete.
    """
    data = []
    folders = [f for f in os.listdir(folder_root) if os.path.isdir(os.path.join(folder_root, f))]

    for folder in tqdm(folders, desc="Scanning folders"):
        full_path = os.path.join(folder_root, folder)
        files = [item for item in os.listdir(full_path) if os.path.isfile(os.path.join(full_path, item))]
        data.append({
            "Folder": folder,
            "Full_Path": full_path,
            "File_Count": len(files)
        })

    df_counts = pd.DataFrame(data)
    df_empty = df_counts[df_counts["File_Count"] == 0]

    print("\n🧾 Total first-level folders:", len(df_counts))
    print("❌ Empty folders found:", len(df_empty))
    print(df_empty[["Folder", "Full_Path"]].to_string(index=False))

    if not df_empty.empty:
        input("\n🟡 Press ENTER to delete the above empty folders...")
        for _, row in df_empty.iterrows():
            try:
                os.rmdir(row["Full_Path"])
                print(f"✅ Deleted: {row['Full_Path']}")
            except Exception as e:
                print(f"⚠️ Error deleting {row['Full_Path']}: {e}")
    else:
        print("\n✅ No empty folders to delete.")

    return df_counts
folder_root = "/Users/yerik/Music/_0_OLD_SOURCE"
df_counts = _paths_2207_i2_GET_delete_empty_firstlevel(folder_root)
```

```python
# -----######-----###### MOVE FILES (< N PER FOLDER) INTO BUCKETS OF 30 FILES MAX -----######-----###### #
import os
import shutil
from tqdm import tqdm
from datetime import datetime

def _paths_2207_i9_BUCKETMOVE_lessNfiles_to_buckets(
    folder_root,
    valid_exts,
    existing_buckets=2,
    max_files_per_bucket=30,
    max_source_file_threshold=19
):
    """
    Moves files from folders with < max_source_file_threshold valid files into numbered __tiny_X_<date> buckets.
    Each bucket gets max_files_per_bucket files. Skips name conflicts and avoids erasing existing buckets.
    """
    today_tag = "07-25"  # Use your own tag if needed
    current_bucket_num = existing_buckets + 1
    bucket_base_path = lambda num: os.path.join(folder_root, f"__tiny_{num}_{today_tag}")
    current_bucket_path = bucket_base_path(current_bucket_num)
    os.makedirs(current_bucket_path, exist_ok=True)
    
    current_bucket_count = len([
        f for f in os.listdir(current_bucket_path)
        if os.path.isfile(os.path.join(current_bucket_path, f))
    ])

    moved_files = []
    skipped_due_to_name = []
    skipped_due_to_size = []
    folders_to_check = []

    folders = [
        os.path.join(folder_root, f)
        for f in os.listdir(folder_root)
        if os.path.isdir(os.path.join(folder_root, f)) and not f.startswith("__")
    ]

    for folder in tqdm(folders, desc="📦 Moving files into buckets"):
        valid_files = []
        for root, _, files in os.walk(folder):
            for file in files:
                if file.startswith("."):
                    continue
                if any(file.lower().endswith(ext) for ext in valid_exts):
                    valid_files.append(os.path.join(root, file))

        if len(valid_files) == 0:
            continue

        if len(valid_files) > max_source_file_threshold:
            skipped_due_to_size.append((folder, len(valid_files)))
            continue

        folders_to_check.append(folder)

        for file_path in valid_files:
            file_name = os.path.basename(file_path)

            # Check for filename conflict
            if os.path.exists(os.path.join(current_bucket_path, file_name)):
                skipped_due_to_name.append(file_name)
                continue

            # Bucket full → rotate
            if current_bucket_count >= max_files_per_bucket:
                current_bucket_num += 1
                current_bucket_path = bucket_base_path(current_bucket_num)
                os.makedirs(current_bucket_path, exist_ok=True)
                current_bucket_count = 0

            dest_path = os.path.join(current_bucket_path, file_name)
            try:
                shutil.move(file_path, dest_path)
                moved_files.append(dest_path)
                current_bucket_count += 1
            except Exception as e:
                print(f"⚠️ Error moving {file_path}: {e}")

    # ----- SAFETY DELETE OR RESCUE EMPTY/NON-EMPTY FOLDERS -----
    empty_folders = []
    check_folder = os.path.expanduser("~/Desktop/__FOLDERS_TO_CHECK_BEFORE_DELETE")
    os.makedirs(check_folder, exist_ok=True)

    for f in folders_to_check:
        remaining_files = []
        for root, _, files in os.walk(f):
            for file in files:
                if file.startswith("."):
                    continue
                remaining_files.append(os.path.join(root, file))

        if remaining_files:
            try:
                dest = os.path.join(check_folder, os.path.basename(f))
                shutil.move(f, dest)
                print(f"📦 Moved non-empty folder to review: {f} → {dest}")
            except Exception as e:
                print(f"⚠️ Error moving {f} to check folder: {e}")
        else:
            empty_folders.append(f)

    if empty_folders:
        print(f"\n🧹 {len(empty_folders)} folders confirmed fully empty. Example:")
        for f in empty_folders[:5]:
            print(f"🗑️  {f}")
        input("\n⏳ Press ENTER to delete these empty folders (or CTRL+C to cancel)...")
        for f in empty_folders:
            try:
                shutil.rmtree(f)
                print(f"✅ Deleted {f}")
            except Exception as e:
                print(f"❌ Error deleting {f}: {e}")
    else:
        print("✅ No empty folders to delete.")

    print(f"\n✅ Moved {len(moved_files)} files across {current_bucket_num - existing_buckets} new buckets.")
    print(f"⛔ Skipped {len(skipped_due_to_size)} folders with too many files")
    print(f"⚠️ Skipped {len(skipped_due_to_name)} files due to name conflict")

    return moved_files
```
