```python
# ----------------------------######----------------------------#
#       _juice_combo_bars_GET_aiff_summary_chunks             #
# ----------------------------######----------------------------#

import os
import librosa
import numpy as np
import pandas as pd
import random
from tqdm import tqdm
import soundfile as sf
from mutagen.aiff import AIFF

# ----------------------------######----------------------------#
def _write_aiff_chunk(y, sr, out_path_aiff, tags_src_path):
    os.makedirs(os.path.dirname(out_path_aiff), exist_ok=True)

    if y.ndim == 1:
        y = np.stack([y, y], axis=0)
    elif y.shape[0] == 1:
        y = np.repeat(y, 2, axis=0)

    y = y.T
    sf.write(out_path_aiff, y, sr, subtype='PCM_16')

    try:
        src_tags = AIFF(tags_src_path)
        dst_tags = AIFF(out_path_aiff)
        if src_tags.tags:
            for tag in src_tags.tags.keys():
                try:
                    dst_tags.tags[tag] = src_tags.tags[tag]
                except Exception:
                    continue
            dst_tags.save()
    except Exception as e:
        print(f"⚠️ Metadata copy failed: {e}")

# ----------------------------######----------------------------#
def _is_chunk_loopable(y_chunk, sr, beat_duration, tolerance_db=3.0):
    head = y_chunk[:, :int(0.1 * sr)]
    tail = y_chunk[:, -int(0.1 * sr):]
    rms_head = np.sqrt(np.mean(head**2))
    rms_tail = np.sqrt(np.mean(tail**2))
    if rms_head == 0 or rms_tail == 0:
        return False
    diff_db = abs(20 * np.log10(rms_head / rms_tail))
    return diff_db <= tolerance_db

# ----------------------------######----------------------------#
def _juice_combo_bars_GET_aiff_summary_chunks(df, lag_sec_input=None):
    print("\n🔁 TQM: Extracting JUICY AIFF chunks (combo mode: beat+bars)...\n")
    chunk_data = []

    bar_weights = {1: 8, 2: 4, 4: 4, 8: 3, 12: 2, 16: 1, 24: 1, 32: 1}
    precision_beats = [1, 2, 4]

    for idx, row in tqdm(df.iterrows(), total=len(df), desc="🌊 Processing Tracks"):
        file_path = row['Path']
        if not file_path.lower().endswith('.aiff'):
            continue

        try:
            bpm = float(row['dominant_bpm'])
            beat_duration = 60 / bpm
            base_name = os.path.splitext(os.path.basename(file_path))[0]

            # Extract _RELEASE_ name from TRkw__RELEASE__ARkw pattern
            try:
                release_name = base_name.split('TRkw__')[1].split('__ARkw')[0]
            except:
                release_name = base_name

            # Base folder name pattern
            folder_name = f"_{release_name}_chopN01_{int(bpm)}"
            out_folder = os.path.join(os.path.dirname(file_path), folder_name)

            y, sr = librosa.load(file_path, sr=None, mono=False)
            if y.ndim == 1:
                y = np.expand_dims(y, axis=0)
            duration_audio = y.shape[1] / sr

            onset_env = librosa.onset.onset_strength(y=np.mean(y, axis=0), sr=sr)
            onset_times = librosa.times_like(onset_env, sr=sr)

            lag_sec = float(lag_sec_input) if lag_sec_input else 0.0

            beat_counter = 0
            bar_counter = 0

            for bars, qty in bar_weights.items():
                chunk_len_sec = beat_duration * bars
                for i in range(qty):
                    if bars in precision_beats:
                        total_region_count = qty
                        segment_length = max((duration_audio - lag_sec) / total_region_count, chunk_len_sec)
                        region_start = lag_sec + (i * segment_length)
                        region_end = min(region_start + segment_length, duration_audio - chunk_len_sec)

                        region_mask = (onset_times >= region_start) & (onset_times < region_end)
                        if not np.any(region_mask):
                            continue

                        region_onsets = onset_env[region_mask]
                        region_times = onset_times[region_mask]
                        sorted_indices = np.argsort(region_onsets)[::-1]

                        for best_idx in sorted_indices:
                            best_time = region_times[best_idx]
                            start = int(best_time * sr)
                            end = start + int(chunk_len_sec * sr)

                            if end > y.shape[1]:
                                pad_len = end - y.shape[1]
                                chunk = y[:, start:]
                                chunk = np.pad(chunk, ((0, 0), (0, pad_len)), mode='constant')
                            else:
                                chunk = y[:, start:end]

                            if _is_chunk_loopable(chunk, sr, beat_duration):
                                out_name = f"beat{beat_counter+1}_{bars}b_{base_name}.aiff"
                                out_path = os.path.join(out_folder, out_name)
                                _write_aiff_chunk(chunk, sr, out_path, tags_src_path=file_path)
                                chunk_data.append({'Path': out_path})
                                beat_counter += 1
                                break
                    else:
                        chunk_samples = int(chunk_len_sec * sr)
                        max_start = y.shape[1] - chunk_samples
                        if max_start <= 0:
                            continue

                        for _ in range(10):
                            start = random.randint(0, max_start)
                            end = start + chunk_samples
                            y_chunk = y[:, start:end]

                            rms = np.sqrt(np.mean(np.mean(y_chunk, axis=0) ** 2))
                            if rms == 0:
                                continue
                            db = 20 * np.log10(rms)
                            if db < -40:
                                continue

                            if _is_chunk_loopable(y_chunk, sr, beat_duration):
                                suffix = f"_bar{bars}_{bar_counter+1}_"
                                out_name = f"{suffix}{base_name}.aiff"
                                out_path = os.path.join(out_folder, out_name)
                                _write_aiff_chunk(y_chunk, sr, out_path, tags_src_path=file_path)
                                chunk_data.append({'Path': out_path})
                                bar_counter += 1
                                break

        except Exception as e:
            print(f"❌ Error with file {file_path}: {e}")
            continue

    df_chunks = pd.DataFrame(chunk_data)
    print(f"\n✅ TQM: Extracted {len(df_chunks)} total loopable chunks.\n")
    return df_chunks
```
