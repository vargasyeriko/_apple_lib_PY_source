```python
# ------------------------------------------------------------
# ########## DARK MOTOWN CHORD GENERATOR ##########
# ------------------------------------------------------------

import pretty_midi


def _chords_0503_darkmotown_GET_midi(output_path):

    pm = pretty_midi.PrettyMIDI()
    piano = pretty_midi.Instrument(program=pretty_midi.instrument_name_to_program('Electric Piano 1'))

    # Dark Motown progression (Detroit soul vibe)
    # Key: D minor

    chords = [
        [50, 57, 60, 64],     # Dm7
        [48, 55, 58, 62],     # Bbmaj7
        [45, 52, 55, 60],     # F7
        [43, 50, 53, 57],     # Gm7
    ]

    duration = 2.0
    start = 0

    for chord in chords:
        for note in chord:
            note_obj = pretty_midi.Note(
                velocity=90,
                pitch=note,
                start=start,
                end=start + duration
            )
            piano.notes.append(note_obj)

        start += duration

    pm.instruments.append(piano)

    pm.write(output_path)

    return output_path
```
