# 0 : CUT 5 KICKS

```python
import pandas as pd

from _fn0_cut_kick import (
    _kick_1011_where_GET_df_kicks,
    _kick_1011_proc_GET_kick_slices,
    _kick_1011_write_GET_kicks,
)

# ====== USER INPUTS ======

mp3_path = "/Users/yerik/Downloads/__DRUM_SAMPLE_TEST/yes.mp3"
out_root = "/Users/yerik/Downloads/__DRUM_SAMPLE_TEST/_KICKS_OUT"
col_path = "Path"

# Build df of source loops
df = pd.DataFrame({col_path: [mp3_path]})

# ====== 1) LOCATE BEST (<=5) KICKS ======
df_kicks = _kick_1011_where_GET_df_kicks(
    df,
    col_path=col_path,
    max_kicks=5,
    low_max_hz=180,
    hop_length=512,
    win_length=2048,
    min_kick_distance_ms=40,
    tqm_bar=True,
)

# ====== 2) PROCESS / CLEAN / NORMALIZE KICKS ======
kick_slices_dict = _kick_1011_proc_GET_kick_slices(
    df_kicks,
    col_path=col_path,
    col_kick_times="kick_times_sec",
    sr_target=44100,
    pre_ms=10.0,
    post_ms=220.0,
    fade_ms=5.0,
    noise_floor_ratio=0.02,
    tail_extra_ms=10.0,
    norm_target_peak=0.99,
    tqm_bar=True,
)

# ====== 3) WRITE KICKS TO DISK ======
exported_kicks = _kick_1011_write_GET_kicks(
    df_kicks,
    kick_slices_dict,
    out_root=out_root,
    col_path=col_path,
    sr_write=44100,
    txt_suffix="_kick_paths.txt",
    tqm_bar=True,
)

print("\n[done] exported kicks:")
for p in exported_kicks:
    print(p)
```

```python
# ===================== 0_FNS =====================
import os

# -----######----- CORE FUNCTION -----######-----
def _exec_1011_generatekick_GET_run():
    print("Press ENTER to execute _fn2_generate_kick.py (or type 'n' to cancel).")
    ans = input("→ ")

    if ans.strip().lower() in ["", "y", "yes"]:
        script = "_fn2_generate_kick.py"

        if not os.path.exists(script):
            print(f"❌ {script} not found in: {os.getcwd()}")
            return

        code = open(script, "r", encoding="utf-8").read()
        # run it as if: python _fn2_generate_kick.py
        exec(compile(code, script, "exec"), {"__name__": "__main__", "__file__": script})
        print(f"✅ Finished executing {script}")
    else:
        print("⚠️ Execution cancelled.")


#!#!#!#!#! RUNNING STATEMENTS !#!#!#!#!#
if __name__ == "__main__":
    _exec_1011_generatekick_GET_run()
```
