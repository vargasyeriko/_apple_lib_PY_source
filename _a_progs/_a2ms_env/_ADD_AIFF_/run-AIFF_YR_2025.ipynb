```python
# -----######-----######-----###### FUNCTION -----######-----######-----######
def _write_tags_2712_id3_SET_tag_bulk(df, path_col, comment_col, genre_col):
    """
    Writes comments and genres to the ID3 tags of AIFF files based on a DataFrame.

    Parameters:
    - df: DataFrame containing file paths, comments, and genres.
    - path_col: Column name in the DataFrame containing AIFF file paths.
    - comment_col: Column name in the DataFrame containing comments to write.
    - genre_col: Column name in the DataFrame containing genres to write.

    Returns:
    - DataFrame: Updated with a 'status' column indicating success or error messages.
    """
    import mutagen
    from mutagen.aiff import AIFF
    from mutagen.id3 import ID3, COMM, TCON

    results = []

    for index, row in df.iterrows():
        aiff_path = row[path_col]
        comment_text = row[comment_col]
        genre_text = row[genre_col]

        try:
            # Load the AIFF file
            audio = AIFF(aiff_path)

            # Add ID3 tag if not already present
            if not hasattr(audio, "tags") or audio.tags is None:
                audio.add_tags()

            # Remove existing comments and genre tags
            if "COMM" in audio.tags:
                del audio.tags["COMM"]
            if "TCON" in audio.tags:
                del audio.tags["TCON"]

            # Add new comment
            audio.tags.add(COMM(encoding=3, lang='eng', desc='', text=comment_text))

            # Add new genre
            audio.tags.add(TCON(encoding=3, text=genre_text))

            # Save changes
            audio.save()

            results.append("Success: Tags updated")
        except Exception as e:
            results.append(f"Error: {e}")

    # Add results to DataFrame
    df["status"] = results
    return df

df =_write_tags_2712_id3_SET_tag_bulk(df, "Path", "comment","genre")
```

```python
# -----######-----###### MAIN FUNCTION -----######-----###### 

def _replace_missing_1212_robust_GET_df(df, cols_1, default="MISSING"):
    """
    Robustly replaces missing values in specified columns using values from the next column
    or a default value when the next column also has missing data.

    Args:
        df: DataFrame to process.
        cols_1: List of columns to align missing values sequentially.
        default: Default value to use when both the current and next column have missing data (default: 'MISSING').

    Returns:
        Updated DataFrame with consistent missing data indicators.
    """
    import numpy as np
    import pandas as pd

    # Define missing indicators explicitly
    missing_indicators = [None, "NaN", "nan", "none", "NaT", "Unknown", "UNKNOWN", np.nan]

    # Ensure all specified columns exist in the DataFrame
    cols_to_process = [col for col in cols_1 if col in df.columns]

    # Convert all specified columns to string type to standardize processing
    df[cols_to_process] = df[cols_to_process].astype(str).replace(missing_indicators, np.nan)

    # Loop through the columns in sequential order
    for i in range(len(cols_to_process) - 1):
        current_col = cols_to_process[i]
        next_col = cols_to_process[i + 1]

        # Fill missing values in the current column using the next column's values
        df[current_col] = df[current_col].fillna(df[next_col])

    # Replace any remaining missing values with the default value
    for col in cols_to_process:
        df[col] = df[col].fillna(default)

    return df

# Define the columns to process
cols_1 = ['title', 'artist', 'rework', 'album', 'genre', 'LABEL',
          'rel_date', 'year', 'rel_year', 'rel_month', 'rel_day', 'BPM', 'KEY']

# Apply the function
df = _replace_missing_1212_robust_GET_df(df, cols_1)
```

```python
# -----######-----###### MAIN FUNCTION -----######-----######

import re
import pandas as pd

def _extract_attrs_1212_DFkw_APPEND_df_results(df, file_col='file_name', missing_keywords=["MISSING", None]):
    """
    Extract attributes from file names in a DataFrame, and fill missing data in specified columns.

    Args:
        df: DataFrame to process.
        file_col: Column containing file names to extract attributes from.
        missing_keywords: List of values considered as missing (default: ["MISSING", None]).

    Returns:
        Updated DataFrame with extracted attributes appended and missing data filled.
    """
    # Define columns to extract
    cols_1 = ['title', 'artist', 'rework', 'album', 'genre', 'LABEL',
              'rel_date', 'year', 'rel_year', 'rel_month', 'rel_day', 'BPM', 'KEY']
    
    # Define regex patterns matching the provided file name structure
    kw_patterns = {
        'title': r'TRkw_(.*?)_ARkw',
        'artist': r'ARkw_(.*?)_MXkw',
        'rework': r'MXkw_(.*?)_KYkw',
        'KEY': r'KYkw_(.*?)_BPkw',
        'BPM': r'BPkw_(\d+)_GNkw',
        'genre': r'GNkw_(.*?)_RMkw',
        'rework': r'RMkw_(.*?)_LBkw',
        'LABEL': r'LBkw_(.*?)_RYkw',
        'album': r'LBkw_(.*?)_RYkw',  # Differentiated from label
        'rel_date': r'RYkw_(\d{4}_\d{2}_\d{2})_PYkw',
        'year': r'RYkw_(\d{4})_\d{2}_\d{2}',
        'rel_year': r'RYkw_(\d{4})_\d{2}_\d{2}',
        'rel_month': r'RYkw_\d{4}_(\d{2})_\d{2}',
        'rel_day': r'RYkw_\d{4}_\d{2}_(\d{2})',
    }
    
    # Process each row to extract attributes
    for index, row in df.iterrows():
        file_name = row[file_col]
        for col in cols_1:
            # Skip if the current value is not missing
            if row[col] not in missing_keywords:
                continue
            
            # Extract attribute using regex pattern
            pattern = kw_patterns.get(col)
            if pattern:
                match = re.search(pattern, file_name)
                if match:
                    # Assign the extracted value to the column
                    df.at[index, col] = match.group(1)
    
    return df
```

```python
import os
import numpy as np
import pandas as pd
import wave
from tqdm import tqdm
from pydub import AudioSegment
import shutil  # To delete the temp folder after all processing
import aubio
import math

# -----######-----###### CORE FUNCTION FOR CONVERSION WITH EXTRA ATTRIBUTES -----######-----######
def _convert_0110_files_to_wav_LOCATE_get_files(df, temp_wav_folder, audio_extensions):
    """
    Converts audio files to WAV if necessary and saves them in the temp_wav_folder.
    Also extracts sample rate (sr_conv), bit depth (bit_depth_conv), and channels (channels_conv) for the converted WAV files.

    :param df: DataFrame containing 'Path' column
    :param temp_wav_folder: The folder to save converted wav files
    :param audio_extensions: List of valid audio extensions (input as a variable outside the function)
    :return: Updated DataFrame with 'Wav_Path', 'sr_conv', 'bit_depth_conv', and 'channels_conv' columns
    """
    
    if not os.path.exists(temp_wav_folder):
        os.makedirs(temp_wav_folder)

    df.loc[:, 'Wav_Path'] = ""
    df.loc[:, 'sr_conv'] = None
    df.loc[:, 'bit_depth_conv'] = None
    df.loc[:, 'channels_conv'] = None

    for index, row in tqdm(df.iterrows(), total=df.shape[0], leave=False, desc="Converting files"):
        file_path = row['Path']
        ext = os.path.splitext(file_path)[1].lower()

        if os.path.exists(file_path) and ext in audio_extensions:
            wav_file_name = os.path.basename(os.path.splitext(file_path)[0] + '.wav')
            wav_path = os.path.join(temp_wav_folder, wav_file_name)

            try:
                # Convert non-WAV files to WAV using pydub
                if ext != '.wav':
                    audio = AudioSegment.from_file(file_path)
                    audio.export(wav_path, format="wav")
                else:
                    wav_path = file_path

                # Extract attributes from the WAV file (for converted file)
                with wave.open(wav_path, 'r') as wav_file:
                    sr_conv = wav_file.getframerate()
                    bit_depth_conv = wav_file.getsampwidth() * 8  # sample width in bytes, so multiply by 8 for bits
                    channels_conv = wav_file.getnchannels()

                # Store the attributes of the converted file in the DataFrame
                df.loc[index, 'Wav_Path'] = wav_path
                df.loc[index, 'sr_conv'] = sr_conv
                df.loc[index, 'bit_depth_conv'] = bit_depth_conv
                df.loc[index, 'channels_conv'] = channels_conv

            except Exception as e:
                print(f"Error processing {file_path}: {e}")

    return df

# -----######-----###### BPM FUNCTION USING AUBIO -----######-----######
def compute_bpm(filename):
    """
    Computes BPM from an audio file using the aubio library.

    :param filename: Path to the audio file
    :return: BPM value (float)
    """
    try:
        win_s = 512  # fft size
        hop_s = win_s // 2  # hop size
        samplerate = 0
        s = aubio.source(filename, samplerate, hop_s)
        samplerate = s.samplerate
        o = aubio.tempo("default", win_s, hop_s, samplerate)
        beats = []
        while True:
            samples, read = s()
            is_beat = o(samples)
            if is_beat:
                this_beat = o.get_last_s()
                beats.append(this_beat)
            if read < hop_s:
                break
        bpms = 60. / np.diff(beats)
        return np.median(bpms) if len(bpms) > 0 else None
    except Exception as e:
        print(f"Error processing file {filename}: {str(e)}")
        return None

# -----######-----###### MAIN FUNCTION TO RUN CONVERSION AND BPM CALCULATION WITH TQM AND CLEANUP -----######-----######
def process_audio_files(df, temp_wav_folder, audio_extensions, chunk_size=1, save_pkl=False, save_path=None):
    """
    Processes audio files in chunks, converts them to WAV if needed, and computes BPM.
    Cleans up the temp folder by deleting processed WAV files after each chunk.
    Optionally saves the DataFrame after processing each chunk to a pickle file.

    :param df: DataFrame containing file paths in 'Path'
    :param temp_wav_folder: Folder to store WAV files
    :param audio_extensions: List of valid audio extensions (input outside the function)
    :param chunk_size: Number of files to process per chunk
    :param save_pkl: Boolean flag to save the DataFrame as a pickle file (default=False)
    :param save_path: Path where the DataFrame is saved after each chunk if save_pkl is True
    :return: Final processed DataFrame
    """
    n = df.shape[0]
    final_df = pd.DataFrame()

    with tqdm(total=n, leave=True, desc="Processing audio files with TQM") as pbar:
        for i in range(0, n, chunk_size):
            df_chunk = df.iloc[i:i + chunk_size].copy()  # Ensuring we are working on a copy to avoid warnings

            # Convert audio files to WAV if needed, and extract sr_conv, bit_depth_conv, channels_conv
            df_chunk = _convert_0110_files_to_wav_LOCATE_get_files(df_chunk, temp_wav_folder, audio_extensions)

            # Compute BPM for each file in the chunk
            df_chunk['ms_bpm'] = [compute_bpm(x) if x.endswith(tuple(audio_extensions)) else None 
                                  for x in tqdm(df_chunk['Wav_Path'], desc="Computing BPM")]

            # Round up the BPM values and convert to integer
            df_chunk['ms_BPM_code'] = df_chunk['ms_bpm'].apply(lambda x: math.ceil(x) if x is not None else None)

            # Add leading zero if BPM is two digits
            df_chunk['ms_BPM_code'] = df_chunk['ms_BPM_code'].apply(lambda x: f"{x:03d}" if pd.notnull(x) else None)

            # Concatenate the results
            final_df = pd.concat([final_df, df_chunk], ignore_index=True)
            
            # Update the main progress bar
            pbar.update(chunk_size)

            # Save after each chunk if save_pkl is True
            if save_pkl and save_path:
                final_df.to_pickle(save_path)
                print(f"Progress saved to {save_path} after chunk {i // chunk_size + 1}")

            # Cleanup: Remove the WAV files created for this chunk
            for file in os.listdir(temp_wav_folder):
                file_path = os.path.join(temp_wav_folder, file)
                if os.path.isfile(file_path) and file.endswith('.wav'):
                    os.remove(file_path)
                    
        # Optionally: Delete the temp folder after all processing
        if os.path.exists(temp_wav_folder):
            shutil.rmtree(temp_wav_folder)

    return final_df

# -----######-----###### EXAMPLE RUN STATEMENTS -----######-----######
# Example DataFrame with paths
audio_extensions = ['.mp3', '.flac', '.aac', '.ogg', '.wma', '.au', '.aif', '.aiff', '.wav']

# Process the files with a temporary WAV folder and save results to pickle if needed
temp_wav_folder = '/Users/yerik/Desktop/temp_wav'
df = process_audio_files(df, temp_wav_folder, audio_extensions, chunk_size=10, save_pkl=False)

df
```

```python
import pandas as pd
import numpy as np
from collections import Counter

# Define the mapping of frequency to MIDI note
def freq_to_midi(frequency):
    return 69 + 12 * np.log2(frequency / 440.0)

# Define the mapping of MIDI note to octave and note name
def midi_to_note_octave(midi_number):
    note_names = ['C', 'C#', 'D', 'D#', 'E', 'F', 'F#', 'G', 'G#', 'A', 'A#', 'B']
    octave = (midi_number // 12) - 1
    note = note_names[int(midi_number % 12)]
    return note, octave

# Function to convert frequency to note and octave
def freq_to_note_octave(frequency):
    midi_number = freq_to_midi(frequency)
    note, octave = midi_to_note_octave(round(midi_number))
    return note, octave

# Camelot wheel mapping
camelot_wheel = {
    'C': '8B', 'C#': '3B', 'D': '10B', 'D#': '5B', 'E': '12B', 'F': '7B', 'F#': '2B', 'G': '9B', 'G#': '4B', 'A': '11B', 'A#': '6B', 'B': '1B',
    'Cm': '5A', 'C#m': '12A', 'Dm': '7A', 'D#m': '2A', 'Em': '9A', 'Fm': '4A', 'F#m': '11A', 'Gm': '6A', 'G#m': '1A', 'Am': '8A', 'A#m': '3A', 'Bm': '10A'
}

# Function to convert note and octave to Camelot key
def note_octave_to_camelot(note, octave):
    key = note + ('m' if octave % 2 == 0 else '')
    return camelot_wheel.get(key, '')

# Convert frequencies to notes and octaves
df['Notes_Octaves'] = df['Prominent_Frequencies'].apply(lambda freqs: [freq_to_note_octave(freq) for freq in freqs])

# Extract first frequency's octave
df['First_Octave'] = df['Prominent_Frequencies'].apply(lambda freqs: freq_to_note_octave(freqs[0])[1])

# Create new string column "octaves" with the format "232"
df['Octaves'] = df['First_Octave'].apply(lambda x: f"{x}{x+1}{x+2}")

# Convert frequencies to notes, octaves, and Camelot keys
df['Notes_Octaves_Camelot'] = df['Prominent_Frequencies'].apply(
    lambda freqs: [note_octave_to_camelot(*freq_to_note_octave(freq)) for freq in freqs]
)

# Extract the Camelot keys and join the first three into a single string
df['Camelot_Sequence'] = df['Notes_Octaves_Camelot'].apply(
    lambda keys: ''.join(keys[:3])
)

# Determine the most repeated Camelot key and its count
def most_repeated_key(keys):
    if keys:
        most_common_key, count = Counter(keys).most_common(1)[0]
        return f"{most_common_key}_{count}"
    return ''

df['key_1'] = df['Notes_Octaves_Camelot'].apply(most_repeated_key)

# KEY NAME CODE
df['key_code'] = (df['key_1'].astype(str) +'cnt-rnk_'+df['Camelot_Sequence'].astype(str) +'-'+df['Octaves'].astype(str)+'oct')

# Chroma cross-check function (optional)
def chroma_cross_check(prominent_freqs):
    # Placeholder: you can run a lightweight chroma analysis or compare harmonic relationships
    top_freq = prominent_freqs[0]  # Simpler: pick the first, or choose based on logic
    return top_freq  # Adjust or improve based on chroma if necessary

df['key_confirmed'] = df['Prominent_Frequencies'].apply(chroma_cross_check)

df
```

```python
import os
import numpy as np
import pandas as pd
import wave
from tqdm import tqdm
from pydub import AudioSegment

# -----######-----###### CORE FUNCTION FOR CONVERSION WITH EXTRA ATTRIBUTES -----######-----######
def _convert_0110_files_to_wav_LOCATE_get_files(df, temp_wav_folder, audio_extensions):
    """
    Converts audio files to WAV if necessary and saves them in the temp_wav_folder.
    Also extracts sample rate (sr_conv), bit depth (bit_depth_conv), and channels (channels_conv) for the converted WAV files.

    :param df: DataFrame containing 'Path' column
    :param temp_wav_folder: The folder to save converted wav files
    :param audio_extensions: List of valid audio extensions (input as a variable outside the function)
    :return: Updated DataFrame with 'Wav_Path', 'sr_conv', 'bit_depth_conv', and 'channels_conv' columns
    """
    
    if not os.path.exists(temp_wav_folder):
        os.makedirs(temp_wav_folder)

    df.loc[:, 'Wav_Path'] = ""
    df.loc[:, 'sr_conv'] = None
    df.loc[:, 'bit_depth_conv'] = None
    df.loc[:, 'channels_conv'] = None

    for index, row in tqdm(df.iterrows(), total=df.shape[0], leave=False, desc="Converting files"):
        file_path = row['Path']
        ext = os.path.splitext(file_path)[1].lower()

        if os.path.exists(file_path) and ext in audio_extensions:
            wav_file_name = os.path.basename(os.path.splitext(file_path)[0] + '.wav')
            wav_path = os.path.join(temp_wav_folder, wav_file_name)

            try:
                # Convert non-WAV files to WAV using pydub
                if ext != '.wav':
                    audio = AudioSegment.from_file(file_path)
                    audio.export(wav_path, format="wav")
                else:
                    wav_path = file_path

                # Extract attributes from the WAV file (for converted file)
                with wave.open(wav_path, 'r') as wav_file:
                    sr_conv = wav_file.getframerate()
                    bit_depth_conv = wav_file.getsampwidth() * 8  # sample width in bytes, so multiply by 8 for bits
                    channels_conv = wav_file.getnchannels()

                # Store the attributes of the converted file in the DataFrame
                df.loc[index, 'Wav_Path'] = wav_path
                df.loc[index, 'sr_conv'] = sr_conv
                df.loc[index, 'bit_depth_conv'] = bit_depth_conv
                df.loc[index, 'channels_conv'] = channels_conv

            except Exception as e:
                print(f"Error processing {file_path}: {e}")

    return df

# -----######-----###### CORE FUNCTION FOR FREQUENCY ANALYSIS -----######-----######
def _analyze_0110_wav_frequencies_LOCATE_get_prominent(df, save_pkl=False):
    """
    Reads WAV files, extracts low, high, and prominent frequencies, and updates the DataFrame.

    :param df: DataFrame with 'Wav_Path' for wav file paths
    :param save_pkl: Boolean flag to save the DataFrame as a pickle file (default=False)
    :return: DataFrame updated with frequency analysis
    """
    low_freqs, high_freqs, prominent_freqs_list = [], [], []

    for index, row in tqdm(df.iterrows(), total=df.shape[0], leave=False, desc="Analyzing frequencies"):
        file_path = row['Wav_Path']
        if file_path and os.path.exists(file_path):
            low_freq, high_freq, prominent_freqs = read_frequencies_and_prominent_wav(file_path)
            
            low_freqs.append(low_freq)
            high_freqs.append(high_freq)
            prominent_freqs_list.append(list(prominent_freqs) if prominent_freqs is not None else [])

            # Save progress after each file if save_pkl is True
            if save_pkl:
                df.to_pickle(f"{os.path.dirname(file_path)}/processed_freqs_df.pkl")
        else:
            low_freqs.append(None)
            high_freqs.append(None)
            prominent_freqs_list.append([])

    df.loc[:, 'Low_Frequency'] = low_freqs
    df.loc[:, 'High_Frequency'] = high_freqs
    df.loc[:, 'Prominent_Frequencies'] = prominent_freqs_list

    return df

# -----######-----###### SUPPORT FUNCTION FOR FREQUENCY EXTRACTION -----######-----######
def read_frequencies_and_prominent_wav(file_path):
    """
    Extracts frequencies from a WAV file.

    :param file_path: Path to the WAV file
    :return: low_freq, high_freq, prominent_freqs
    """
    with wave.open(file_path, 'r') as f:
        nframes = f.getnframes()
        sample_rate = f.getframerate()
        audio_data = np.frombuffer(f.readframes(nframes), dtype=np.int16)

        fft_values = np.fft.fft(audio_data)
        freq_values = np.fft.fftfreq(len(audio_data), 1/sample_rate)

        fft_values = fft_values[(freq_values >= 10) & (freq_values <= 4000)]
        freq_values = freq_values[(freq_values >= 10) & (freq_values <= 4000)]

        positive_indices = np.where(freq_values > 0)
        positive_freqs = freq_values[positive_indices]
        positive_fft_values = np.abs(fft_values[positive_indices])

        if len(positive_freqs) == 0:
            return None, None, None

        low_freq = np.min(positive_freqs)
        high_freq = np.max(positive_freqs)
        prominent_indices = positive_fft_values.argsort()[-10:][::-1]
        prominent_freqs = positive_freqs[prominent_indices]

    return low_freq, high_freq, prominent_freqs

# -----######-----###### MAIN FUNCTION TO RUN CONVERSION AND ANALYSIS WITH TQM -----######-----######
def process_audio_files(df, temp_wav_folder, audio_extensions, chunk_size=1, save_pkl=False):
    """
    Processes audio files in chunks: converts audio to WAV if needed, then extracts prominent frequencies.
    Optionally saves the DataFrame after processing each file to a pickle file.

    :param df: DataFrame containing file paths in 'Path'
    :param temp_wav_folder: Folder to store WAV files
    :param audio_extensions: List of valid audio extensions (input outside the function)
    :param chunk_size: Number of files to process per chunk
    :param save_pkl: Boolean flag to save the DataFrame as a pickle file (default=False)
    :return: Final processed DataFrame
    """
    n = df.shape[0]
    final_df = pd.DataFrame()

    with tqdm(total=n, leave=True, desc="Processing audio files with TQM") as pbar:
        for i in range(0, n, chunk_size):
            df_chunk = df.iloc[i:i + chunk_size].copy()  # Ensuring we are working on a copy to avoid warnings

            # Convert audio files to WAV if needed, and extract sr_conv, bit_depth_conv, channels_conv
            df_chunk = _convert_0110_files_to_wav_LOCATE_get_files(df_chunk, temp_wav_folder, audio_extensions)

            # Analyze frequencies from WAV files
            df_chunk = _analyze_0110_wav_frequencies_LOCATE_get_prominent(df_chunk, save_pkl)

            # Concatenate the results
            final_df = pd.concat([final_df, df_chunk], ignore_index=True)
            
            pbar.update(chunk_size)  # Update the main progress bar

    return final_df

# -----######-----###### EXAMPLE RUN STATEMENTS -----######-----######
# Define valid audio extensions outside of the functions
audio_extensions = ['.mp3', '.flac', '.aac', '.ogg', '.wma', '.au', '.aif', '.aiff', '.wav']

# Process the files with a temporary WAV folder and save results to pickle if needed
temp_wav_folder = '/Users/yerik/Desktop/temp_wav'
df_p = process_audio_files(df, temp_wav_folder, audio_extensions, chunk_size=1, save_pkl=False)

df_p
```

```python
%%writefile -a /Users/yerik/_apple_source/PY/functions/_GLOBAL/_2ms/_MSG_GET_song_attributes.py

import numpy as np
import pandas as pd
from pydub import AudioSegment
from pyloudnorm import Meter
from tqdm import tqdm

# -----######-----###### LUFS Calculation Function -----######-----######


def _all_au_ext_GET_12_LUFS_values(df):
    """
    Computes LUFS for each file in the DataFrame using the path from 'Path' and updating meter based on sample rate from the DataFrame.
    
    Parameters:
    df (DataFrame): DataFrame containing 'Path', 'sr', and other relevant columns.

    Returns:
    List: List of LUFS values for each file.
    """
    
    lufs_results = []

    # Loop through paths with tqdm progress bar
    for _, row in tqdm(df.iterrows(), desc="Computing LUFS", unit="file", total=len(df)):
        try:
            # Get the file path and sample rate from the DataFrame
            file_path = row['Path']
            sample_rate = row['sr']
            
            # Load the audio file using pydub
            audio = AudioSegment.from_file(file_path)
            
            # Convert audio to an array of samples
            samples = np.array(audio.get_array_of_samples())
            
            # Convert integer samples to floating point and normalize
            samples = samples / (2**15)
            
            # Check if stereo or mono and reshape accordingly
            if audio.channels == 2:
                samples = samples.reshape((-1, 2))
            else:
                samples = samples.reshape((-1, 1))
            
            # Initialize the loudness meter with the sample rate from DataFrame
            meter = Meter(rate=sample_rate)
            
            # Compute LUFS
            loudness = meter.integrated_loudness(samples)
            lufs_results.append(round(loudness, 3))
        
        except Exception as e:
            print(f"Error processing {file_path}: {e}")
            lufs_results.append(None)
    
    return lufs_results

# -----######-----###### LUFS Mapping Function -----######-----######
def _all_values_CREATE_12_LUFS_categories(lufs_value):
    """
    Maps LUFS values to energy levels.

    Parameters:
    lufs_value (float): The LUFS value.

    Returns:
    str: Energy level code corresponding to LUFS value.
    """
    if lufs_value is None:
        return None
    elif lufs_value <= -18:
        return 'LuA'
    elif -18 < lufs_value <= -16:
        return 'LuB'
    elif -16 < lufs_value <= -14:
        return 'LuC'
    elif -14 < lufs_value <= -12:
        return 'LuD'
    elif -12 < lufs_value <= -10:
        return 'Lue'
    else:
        return 'LuF'

```
