```python
import re

def has_empty_title_between_keys(temp_id, front, back):
    if isinstance(temp_id, str) and front in temp_id and back in temp_id:
        start = temp_id.find(front) + len(front)
        end = temp_id.find(back, start)
        content = temp_id[start:end].strip('_ ').lower()
        return content == ''
    return False

# Set your markers
btw_front = 'RMkw'
btw_back = 'LBkw'

# Find where there's nothing in between
df_empty_titles = df[df['temp_id'].apply(lambda x: has_empty_title_between_keys(x, btw_front, btw_back))]
df_empty_titles[['temp_id', 'Path']]
```
