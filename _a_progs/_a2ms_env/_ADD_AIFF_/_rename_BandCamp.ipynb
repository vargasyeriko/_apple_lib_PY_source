```python
# -----######-----###### RENAME FINAL: MXkw + GENRE FIX -----######-----######
import os
import pandas as pd
from tqdm import tqdm
from datetime import datetime

def _rename_0104_kwtagging_GET_renamed_files_from_txt(txt_path):
    tqdm.pandas()

    # Read UTF-16 with proper engine
    df = pd.read_csv(txt_path, sep="\t", encoding='utf-16', engine='python')
    df.columns = df.columns.str.strip()

    def clean(s):
        return (
            str(s)
            .replace(" ", "_")
            .replace("/", "___")
            .replace(",", "_")
            .replace("(", "")
            .replace(")", "")
            .replace("!", "")
            .replace("&", "and")
            .replace("’", "")
            .replace("'", "")
            .replace("¿", "")
            .replace("¡", "")
            .replace(":", "")
            .replace(";", "")
            .strip()
        )

    def extract_mix(title):
        title_lower = title.lower()
        if "remix" in title_lower or "mix" in title_lower:
            return clean(title)
        return "original"

    def format_filename(row):
        title = clean(row.get('Track Title', ''))
        remix = extract_mix(title)
        key = clean(row.get('Key', 'NA')) if pd.notna(row.get('Key', '')) else 'NA'
        bpm = str(int(round(float(row.get('BPM', 0))))) if pd.notna(row.get('BPM', 0)) else 'NA'
        genre = "Mexican_House"
        date_added = pd.to_datetime(row.get('Date Added', datetime.today())).strftime('%Y_%m_%d')
        ext = os.path.splitext(row.get('Location', ''))[1]

        artist = "Salsa_various"
        label = " Sabrosura_Records"
        release_date = "2014_06_28"

        new_name = (
            f"TRkw_{title}_ARkw_{artist}_MXkw_{remix}_KYkw_{key}_BPkw_{bpm}_"
            f"GNkw_{genre}_RMkw__LBkw_{label}_RYkw_{release_date}_PYkw_{date_added}{ext}"
        )
        return new_name

    new_paths = []
    for i, row in tqdm(df.iterrows(), total=len(df), desc="Renaming files"):
        original_path = str(row.get('Location', '')).strip()

        if not os.path.isfile(original_path):
            print(f"❌ File not found (Row {i}): {original_path}")
            new_paths.append(None)
            continue

        new_filename = format_filename(row)
        new_path = os.path.join(os.path.dirname(original_path), new_filename)

        try:
            os.rename(original_path, new_path)
            new_paths.append(new_path)
        except Exception as e:
            print(f"❌ Error renaming {original_path}: {e}")
            new_paths.append(None)

    df['Renamed_Path'] = new_paths
    return df
```
