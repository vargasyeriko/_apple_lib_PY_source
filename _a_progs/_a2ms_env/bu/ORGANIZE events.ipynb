```python
import os
from datetime import datetime

# Subfolders with smart numbering
subfolders = [
    "00_event_flier",
    "01_raw_recording",
    "02_raw_media",
    "03_icloud_share_sets",
    "04_m3u8_playlists",
    "05_short_videos",
    "06_YODJ_upload",
    "07_YODJ_upload_artwork_statement",
    "08_professional_photos",
    "09_sponsors_materials"
]

# Base directory for new event folders
base_dir = "/Users/yerik/Desktop/try"

def create_single_event():
    # Get the current year in short format (e.g., "25" for 2025)
    current_year_suffix = datetime.now().year % 100  # Gets last two digits of the year
    full_year = f"20{current_year_suffix}"  # Convert "25" -> "2025"
    
    # Ensure base directory exists
    os.makedirs(base_dir, exist_ok=True)

    # Ask user for event name
    event_name = input("Enter the name of the new event folder: ").strip()
    if not event_name:
        print("❌ No name entered. Aborting folder creation.")
        return
    
    # Format event folder name with the full year
    event_folder_with_year = f"{full_year}_EV_{event_name}"
    event_path = os.path.join(base_dir, event_folder_with_year)
    
    # Create the main event folder
    os.makedirs(event_path, exist_ok=True)

    # Create subfolders inside the event folder
    for subfolder in subfolders:
        subfolder_path = os.path.join(event_path, subfolder)
        os.makedirs(subfolder_path, exist_ok=True)

    print(f"✅ Created: {event_path} with subfolders {subfolders}")

# Call function (or use it wherever needed)
create_single_event()
```
