```python
# -----######-----###### PATH SCANNER W/ FIRST 5 CHAR ID -----######-----###### #
import os
import pandas as pd
from tqdm import tqdm

def _paths_1807_id5_GET_df_paths_ids(folder_root):
    """
    Extracts all file paths into df['Path'] and assigns ID as the first 5 characters of file_name.
    """
    all_paths = []
    for root, _, files in os.walk(folder_root):
        for file in files:
            if file.startswith('._') or file == '.DS_Store':
                continue
            all_paths.append(os.path.join(root, file))

    df = pd.DataFrame({'Path': all_paths})
    df['file_name'] = df['Path'].apply(os.path.basename)
    df['ID'] = df['file_name'].str[:5]  # First 5 characters

    return df
```

```python
folder_root = "/Volumes/MUSIC_PROD/_4_MUSIC_PROD/_0_PROD_MATERIAL/_ALL_6_PROD_exports/_25-56_EXPORTS__"
df = _paths_1807_id5_GET_df_paths_ids(folder_root)
df
```

```python
# Get sets of unique IDs from both DataFrames
ids_exports = set(df_exports['ID'])
ids_main = set(df['ID'])

# Find intersection (common IDs)
common_ids = ids_exports.intersection(ids_main)

# Print results
print(f"✅ Total IDs in df_exports: {len(ids_exports)}")
print(f"✅ Total IDs in df:         {len(ids_main)}")
print(f"🔁 Matching IDs:            {len(common_ids)}")
```

```python
# Filter df_export to only rows with matching IDs
df_ = df_exports[df_exports['ID'].isin(common_ids)]
```

```python
df_
```

```python
# Drop old Path in df_ (if exists), then merge correct Path from df
df_ = df_.drop(columns=['Path'], errors='ignore')
df_ = df_.merge(df[['ID', 'Path']], on='ID', how='left')
```

```python
print("🎯 Total updated rows in df_:", len(df_))
print("✅ Non-null paths:", df_['Path'].notna().sum())
print("❌ Missing paths:", df_['Path'].isna().sum())
```

```python
import os

# Add 'base_name' column from 'Path'
df_['base_name'] = df_['Path'].apply(lambda x: os.path.basename(x) if pd.notna(x) else None)

# Reorder columns: Path, base_name, ID first — rest stay in original order
cols = ['Path', 'base_name', 'ID'] + [col for col in df_.columns if col not in ['Path', 'base_name', 'ID']]
df_ = df_[cols]
```
