```python
# -----######-----###### DELUXE AIFF NORMALIZER (NO CLIP) -----######-----######
import os
import subprocess
import numpy as np
import soundfile as sf
from tqdm import tqdm

def _audio_0205_g3_GET_deluxe_norm_aiff(folder_path):
    """
    Normalize AIFF files in a folder to -0.1 dBFS safely and precisely.
    Uses numpy + soundfile for high-fidelity normalization.
    """

    if not os.path.isdir(folder_path):
        raise NotADirectoryError(f"❌ Folder does not exist: {folder_path}")

    files = [f for f in os.listdir(folder_path) if f.lower().endswith(".aiff") and not f.startswith("._")]
    print(f"🎧 Found {len(files)} AIFF files. Starting deluxe normalization...\n")

    for filename in tqdm(files, desc="🎚️  Deluxe Normalizing"):
        full_path = os.path.join(folder_path, filename)

        try:
            data, sr = sf.read(full_path, dtype='float32')
            peak = np.max(np.abs(data))

            if peak == 0:
                continue  # silent

            target_peak = 10 ** (-0.1 / 20)  # ≈ 0.98855
            gain = target_peak / peak

            if 0.95 < gain < 1.05:
                continue  # already normalized well

            norm_data = data * gain
            if np.max(np.abs(norm_data)) >= 1.0:
                print(f"⚠️ Potential clip detected in {filename} — skipped.")
                continue

            # Overwrite safely
            sf.write(full_path, norm_data, sr, format='AIFF')

        except Exception as e:
            print(f"❌ Error with {filename}: {e}")
_audio_0205_g3_GET_deluxe_norm_aiff('/Users/yerik/Music/EXPORTS_AIFF_M-PLUS')
```
