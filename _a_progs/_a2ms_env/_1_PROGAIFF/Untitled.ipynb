```python
# -----######-----######-----######-----######-----######-----#
# 0_FNS: _dfcat_1704_batch_GET_df_concat_allpkl
# -----######-----######-----######-----######-----######-----#
import os
import pandas as pd
from tqdm import tqdm

def _dfcat_1704_batch_GET_df_concat_allpkl(folder_path):
    """
    Concatenates all PKL files named df_000.pkl to df_036.pkl in the folder.
    
    Parameters:
        folder_path (str): Directory path containing the .pkl files

    Returns:
        pd.DataFrame: Concatenated DataFrame of all valid .pkl files
    """
    dfs = []
    for i in tqdm(range(37), desc="📦 Loading .pkl files"):
        fname = f"df_{i:03d}_.pkl"
        full_path = os.path.join(folder_path, fname)
        if os.path.exists(full_path):
            try:
                df = pd.read_pickle(full_path)
                dfs.append(df)
            except Exception as e:
                print(f"❌ Error reading {fname}: {e}")
        else:
            print(f"⚠️ File not found: {fname}")
    
    if dfs:
        df_concat = pd.concat(dfs, ignore_index=True)
        print(f"✅ Concatenated {len(dfs)} files. Total rows: {len(df_concat)}")
        return df_concat
    else:
        print("❌ No valid .pkl files loaded.")
        return pd.DataFrame()
# Define your directory path containing df_000.pkl to df_036.pkl
folder_path = ""

# Run the function
df_all = _dfcat_1704_batch_GET_df_concat_allpkl(folder_path)
```

```python
# -----######-----######-----######-----######-----######-----#
# 0_FNS: _cover_1704_i3_improve_embed_GET_aiff_robust
# -----######-----######-----######-----######-----######-----#

import os
from mutagen.aiff import AIFF
from mutagen.id3 import ID3, APIC
from PIL import Image
import io

def _cover_1704_i3_improve_embed_GET_aiff_robust(aiff_path, upscale_factor=3):
    """
    Extracts and upscales cover art in AIFF (even if tag is oddly named).
    Prints resolution before/after and embeds back into AIFF.
    """
    try:
        audio = AIFF(aiff_path)

        # Force-load ID3 if tags seem missing
        if not audio.tags:
            try:
                audio.tags = ID3(aiff_path)
            except Exception as e:
                print(f"❌ No ID3 tags found or readable: {e}")
                return False

        # Find the APIC frame (cover art)
        apic = None
        for key in audio.tags.keys():
            if key.startswith('APIC'):
                apic = audio.tags[key]
                break

        if apic is None:
            print(f"❌ Cover art not found in: {aiff_path}")
            return False

        # Read and upscale image
        img_data = apic.data
        img = Image.open(io.BytesIO(img_data)).convert("RGB")
        w, h = img.size
        print(f"🖼️  Original: {w} x {h} px")

        new_size = (w * upscale_factor, h * upscale_factor)
        img_up = img.resize(new_size, Image.LANCZOS)
        print(f"📈 Upscaled: {new_size[0]} x {new_size[1]} px")

        # Save new image as bytes
        img_byte_arr = io.BytesIO()
        img_up.save(img_byte_arr, format='JPEG')
        img_bytes = img_byte_arr.getvalue()

        # Replace and save
        new_apic = APIC(
            encoding=3,
            mime='image/jpeg',
            type=3,
            desc='Cover',
            data=img_bytes
        )
        audio.tags.delall("APIC")
        audio.tags.add(new_apic)
        audio.save()

        print(f"✅ Updated cover embedded: {os.path.basename(aiff_path)}")
        return True

    except Exception as e:
        print(f"❌ ERROR processing {aiff_path}: {e}")
        return False
```
