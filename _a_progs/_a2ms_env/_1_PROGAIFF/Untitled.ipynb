```python
# -----------------------------######-----------------------------###### 
def _rename_1804_folder_GET_songcount_names(folder_path):  
    """
    Renames all AIFF files in a given folder to song_1.aiff, song_2.aiff, etc.

    Parameters:
        folder_path: Path to the folder containing .aiff files.
        
    Returns:
        Prints renamed files. Overwrites existing names. Skips non-AIFF files.
    """
    import os
    from pathlib import Path

    folder = Path(folder_path)
    aiff_files = sorted([f for f in folder.iterdir() if f.is_file() and f.suffix.lower() == '.aiff'])

    for i, file in enumerate(aiff_files, 1):
        new_name = folder / f"song_{i}.aiff"
        try:
            file.rename(new_name)
            print(f"✅ Renamed: {file.name} ➜ {new_name.name}")
        except Exception as e:
            print(f"❌ Failed: {file.name} ➜ {new_name.name} | Reason: {e}")

_rename_1804_folder_GET_songcount_names("/Users/yerik/Desktop/emergency_newsongs")
```
