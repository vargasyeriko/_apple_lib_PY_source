```python
import os
import imageio_ffmpeg
from mutagen.easyid3 import EasyID3
from pydub import AudioSegment

# 1. Point pydub directly to the bundled FFmpeg binary so Jupyter never loses track of it
AudioSegment.converter = imageio_ffmpeg.get_ffmpeg_exe()


def convert_and_verify(
    input_wav: str,
    output_mp3: str,
    artist_name: str = "JerikODJ",
    track_title: str = "Untitled Track",
):
  if not os.path.exists(input_wav):
    print(f"❌ File not found: '{input_wav}'")
    print(f"   Current folder is: {os.getcwd()}")
    print(
        f"   Files here: {[f for f in os.listdir('.') if f.endswith('.wav')]}"
    )
    return

  # 2. Define ownership tags
  ownership_tags = {
      "artist": artist_name,
      "albumartist": artist_name,
      "title": track_title,
      "copyright": f"© 2026 {artist_name}. All Rights Reserved.",
      "comment": f"Official Master - Owned by {artist_name}",
  }

  # 3. Convert WAV to high-quality 320kbps MP3 with tags
  print(f"⏳ Converting '{input_wav}'...")
  audio = AudioSegment.from_wav(input_wav)
  audio.export(output_mp3, format="mp3", bitrate="320k", tags=ownership_tags)
  print(f"✅ Saved: '{output_mp3}' ({os.path.getsize(output_mp3) / 1024:.1f} KB)")

  # 4. Read the tags back from the saved MP3 to prove they are locked in
  print("\n🔍 Verified Internal ID3 Tags:")
  saved_tags = EasyID3(output_mp3)
  for key, value in saved_tags.items():
    print(f"   • {key.upper()}: {value[0]}")


```
