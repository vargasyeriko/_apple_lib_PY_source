# Spectral Shape-Texture Map

```python
# ---------------------------------
# IMPORTS
# ---------------------------------
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle

# ---------------------------------
# STYLE CONFIG
# ---------------------------------
plt.rcParams.update({
    "font.family": "DejaVu Sans",
    "axes.titleweight": "bold"
})

# ---------------------------------
# CREATE FIGURE
# ---------------------------------
fig, ax = plt.subplots(figsize=(9, 5))

# ---------------------------------
# COLOR MATRIX (NOW 5 COLUMNS)
# ---------------------------------
colors = [
    ['#ADD8FF', '#3EA4FF', '#005EFF', '#113C70', '#000B2D'],  # tonal
    ['#FFE8CF', '#FFC78C', '#FFA64F', '#FF7600', '#C84200']   # noisy
]

labels = [
    ['Sub', 'Groove Low', 'Mix Body', 'Hi-End', 'Shimmer'],
    ['Soft Noise', 'Noisy Groove', 'Distortion', 'Crackle', 'Harsh']
]

# ---------------------------------
# DRAW GRID
# ---------------------------------
for y in range(2):
    for x in range(5):
        rect = Rectangle(
            (x, y), 1, 1,
            facecolor=colors[y][x],
            edgecolor='white',
            linewidth=2
        )
        ax.add_patch(rect)

        # adaptive text color
        text_color = 'black' if x < 2 else 'white'

        ax.text(
            x + 0.5, y + 0.5,
            labels[y][x],
            ha='center', va='center',
            fontsize=10,
            weight='bold',
            color=text_color
        )

# ---------------------------------
# AXES CLEANUP
# ---------------------------------
ax.set_xlim(0, 5)
ax.set_ylim(0, 2)

for spine in ax.spines.values():
    spine.set_visible(False)

# ---------------------------------
# AXIS LABELS (UPDATED)
# ---------------------------------
ax.set_xticks([0.5, 1.5, 2.5, 3.5, 4.5])
ax.set_xticklabels([
    '<2k', '2–3k', '3–4.2k', '4.2–5.2k', '5.2k+'
], fontsize=9)

ax.set_yticks([0.5, 1.5])
ax.set_yticklabels(['Tonal', 'Noisy'], fontsize=10)

# ---------------------------------
# TITLES
# ---------------------------------
ax.set_title('Spectral Shape–Texture Map (Full Resolution)', fontsize=14, pad=15)

ax.set_xlabel('Bandwidth → Frequency Spread (Shape)', fontsize=10, labelpad=10)
ax.set_ylabel('Flatness → Texture (Tone ↔ Noise)', fontsize=10, labelpad=10)

# ---------------------------------
# BACKGROUND
# ---------------------------------
ax.set_facecolor('#f7f7f7')
fig.patch.set_facecolor('white')

plt.tight_layout()
plt.show()
```

# Spectral centroid - where most frequencies sit

```python
# ---------------------------------
# IMPORTS
# ---------------------------------
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle

# ---------------------------------
# STYLE
# ---------------------------------
plt.rcParams.update({
    "font.family": "DejaVu Sans",
    "axes.titleweight": "bold"
})

intervals = spectral_centroid_intervals_1504

# ---------------------------------
# FIGURE (taller for labels)
# ---------------------------------
fig, ax = plt.subplots(figsize=(16, 5))

# ---------------------------------
# DRAW THIN SPECTRUM BAR
# ---------------------------------
bar_height = 0.35   # << thinner blocks (key fix)

for i, (low, high, color, label) in enumerate(intervals):
    rect = Rectangle(
        (i, 0.6), 1, bar_height,
        facecolor=color,
        edgecolor='white',
        linewidth=1
    )
    ax.add_patch(rect)

# ---------------------------------
# LABELS BELOW (NO OVERLAP)
# ---------------------------------
for i, (low, high, color, label) in enumerate(intervals):
    ax.text(
        i + 0.5,
        0.45,
        label,
        ha='right',
        va='top',
        rotation=45,   # << angled for spacing
        fontsize=8
    )

# ---------------------------------
# AXIS CLEANUP
# ---------------------------------
ax.set_xlim(0, len(intervals))
ax.set_ylim(0, 1.2)

for spine in ax.spines.values():
    spine.set_visible(False)

ax.set_yticks([])

# ---------------------------------
# X TICKS (FREQUENCY SCALE)
# ---------------------------------
xticks = list(range(0, len(intervals), 4))
xlabels = [f"{int(intervals[i][0])}" for i in xticks]

ax.set_xticks(xticks)
ax.set_xticklabels(xlabels, fontsize=9)

# ---------------------------------
# TITLES
# ---------------------------------
ax.set_title('Spectral Centroid — Position / Brightness / Emotion', fontsize=14, pad=15)
ax.set_xlabel('Frequency (Hz)', fontsize=11)

# ---------------------------------
# BACKGROUND
# ---------------------------------
ax.set_facecolor('#f7f7f7')
fig.patch.set_facecolor('white')

plt.tight_layout()
plt.show()
```
