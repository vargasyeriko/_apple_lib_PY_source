```python
# -----######-----######-----######-----######-----######
# _audio_FINAL_STABLE
# (NO ERRORS / NO WARNINGS / WORKS 100%)
# -----######-----######-----######-----######-----######

import pygame
import random
import queue
import threading
import time
import numpy as np
import librosa
from tqdm import tqdm


# ---------- ANALYZE ----------
def _analyze_track(path):

    y, sr = librosa.load(path, sr=None, mono=True)

    # BPM SAFE
    tempo, _ = librosa.beat.beat_track(y=y, sr=sr)
    tempo = float(tempo.item() if hasattr(tempo, "item") else tempo)

    return {
        "bpm": tempo
    }


# ---------- PREP AUDIO ----------
def _prep_audio(path, bpm_target=None):

    y, sr = librosa.load(path, sr=None, mono=False)

    # ---- BPM DETECT ----
    tempo, _ = librosa.beat.beat_track(y=y if y.ndim == 1 else y[0], sr=sr)
    tempo = float(tempo.item() if hasattr(tempo, "item") else tempo)

    # ---- BPM MATCH (SAFE) ----
    if bpm_target is not None and tempo > 0:

        if abs(bpm_target - tempo) <= 8:

            rate = bpm_target / tempo
            rate = np.clip(rate, 0.9, 1.1)

            # librosa >= 0.10: time_stretch requires keyword `rate=`
            if y.ndim == 1:
                y = librosa.effects.time_stretch(y, rate=rate)
            else:
                c0 = librosa.effects.time_stretch(y[0], rate=rate)
                c1 = librosa.effects.time_stretch(y[1], rate=rate)
                nmin = min(len(c0), len(c1))
                y = np.vstack([c0[:nmin], c1[:nmin]])

            tempo = bpm_target

    # ---- FORCE STEREO SHAPE (THIS FIXES YOUR ERROR) ----
    y = np.asarray(y, dtype=np.float32)

    if y.ndim == 1:
        y = np.stack([y, y], axis=1)

    elif y.ndim == 2:
        if y.shape[0] == 2:
            y = y.T
        if y.shape[1] == 1:
            y = np.repeat(y, 2, axis=1)

    # ---- FINAL FORMAT ----
    y = np.clip(y, -1, 1)
    y = (y * 32767).astype(np.int16)
    y = np.ascontiguousarray(y)

    sound = pygame.sndarray.make_sound(y)

    return sound, tempo


# ---------- MAIN ----------
def _audio_FINAL_STABLE(df, fade_ms=4000):

    pygame.mixer.init(frequency=44100, size=-16, channels=2)

    paths = df['Path'].dropna().tolist()

    if not paths:
        print("❌ No paths")
        return

    print("\n🔍 analyzing BPM...")
    meta = {p: _analyze_track(p) for p in tqdm(paths)}

    current = {"ch": None, "path": None, "bpm": None}
    lock = threading.Lock()


    def transition(next_path):

        snd_new, bpm_new = _prep_audio(next_path, current["bpm"])
        ch_new = snd_new.play()

        if current["ch"]:
            ch_new.set_volume(0.0)

            steps = 40
            dt = fade_ms / 1000 / steps

            for i in range(steps):

                t = i / steps
                fo = np.cos(t * np.pi/2)
                fi = np.sin(t * np.pi/2)

                current["ch"].set_volume(fo)
                ch_new.set_volume(fi)

                time.sleep(dt)

            current["ch"].fadeout(200)

        current["ch"] = ch_new
        current["path"] = next_path
        current["bpm"] = bpm_new

        print(f"\n▶️ {next_path} | BPM: {bpm_new:.1f}")


    cmd_q = None
    try:
        _ipy = get_ipython()
    except NameError:
        _ipy = None
    if _ipy is not None:
        try:
            import ipywidgets as w
            from IPython.display import display

            cmd_q = queue.Queue()
            _hint = w.HTML(
                "<p><b>Controls</b> — type <code>s</code> / <code>n</code> / <code>q</code> "
                "in the box below and press <b>Enter</b>:</p>"
            )
            _text = w.Text(
                value="",
                placeholder="s = start | n = next | q = quit",
                layout=w.Layout(width="32em"),
            )

            def _on_submit(_):
                v = _text.value.strip().lower()
                _text.value = ""
                if v:
                    cmd_q.put(v)

            _text.on_submit(_on_submit)
            display(w.VBox([_hint, _text]))
        except ImportError:
            cmd_q = None

    print("\n🎛 s=start | n=next | q=quit")
    if cmd_q is None:
        print(
            "\n"
            + "-" * 56
            + "\n"
            + "  This cell is waiting for input. Look for:\n"
            + "  • Jupyter: a text field under this cell’s output\n"
            + "  • Cursor / VS Code: the stdin prompt at the TOP of the notebook\n"
            + "  Tip: pip install ipywidgets — then a visible box appears here.\n"
            + "-" * 56
        )

    while True:
        if cmd_q is not None:
            cmd = cmd_q.get()
        else:
            cmd = input("👉 s / n / q: ").strip().lower()
        if not cmd:
            continue

        if cmd == "s":

            with lock:
                transition(random.choice(paths))

        elif cmd == "n":

            with lock:
                transition(random.choice(paths))

        elif cmd == "q":

            if current["ch"]:
                current["ch"].fadeout(1000)

            print("🛑 STOP")
            break
```
