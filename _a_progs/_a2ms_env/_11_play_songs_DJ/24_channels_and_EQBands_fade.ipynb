```python
# -----######-----######-----######-----######-----######
# _audio_1104_i12_GET_df_player_FINAL_FIXED
# (FIXED: C-CONTIGUOUS + KICK + AUDIO QUALITY + NO CRASH)
# -----######-----######-----######-----######-----######

import pygame
import random
import threading
import time
import numpy as np
import librosa
from tqdm import tqdm


# ---------- ANALYZE ----------
def _analyze_track(path):

    y, sr = librosa.load(path, sr=None, mono=True)

    duration = librosa.get_duration(path=path)

    # 🔥 LOW FREQ ENERGY (KICK DETECTION)
    S = np.abs(librosa.stft(y))
    freqs = librosa.fft_frequencies(sr=sr)

    low = S[(freqs >= 20) & (freqs <= 150)]
    energy = np.mean(low, axis=0)

    peaks = np.where(energy > np.percentile(energy, 96))[0]

    if len(peaks) > 0:
        start_sec = librosa.frames_to_time(peaks[0], sr=sr)
    else:
        start_sec = 0

    return {"duration": duration, "start_sec": start_sec}


# ---------- TRIM ----------
def _trim_to_kick(path, start_sec):

    y, sr = librosa.load(path, sr=None, mono=False)

    if y.ndim == 2:
        y = y[:, int(start_sec * sr):]
    else:
        y = y[int(start_sec * sr):]

    # 🔥 CRITICAL FIX
    y = np.ascontiguousarray(y)

    return y, sr


# ---------- SIMPLE EQ ----------
def _apply_eq(y, sr, low=1.0, mid=1.0, high=1.0):

    S = librosa.stft(y if y.ndim == 1 else y[0])  # safe mono ref
    freqs = librosa.fft_frequencies(sr=sr)

    S[freqs < 200] *= low
    S[(freqs >= 200) & (freqs < 2000)] *= mid
    S[freqs >= 2000] *= high

    y_eq = librosa.istft(S)

    # 🔥 restore stereo if needed
    if y.ndim == 2:
        y_eq = np.column_stack((y_eq, y_eq))

    return y_eq


# ---------- TO SOUND ----------
def _to_sound(y):

    y = np.asarray(y, dtype=np.float32)
    y = np.clip(y, -1, 1)
    y = (y * 32767).astype(np.int16)

    if y.ndim == 1:
        y = np.column_stack((y, y))

    # 🔥 CRITICAL FIX
    y = np.ascontiguousarray(y)

    return pygame.sndarray.make_sound(y)


# ---------- MAIN ----------
def _audio_1104_i12_GET_df_player_FINAL_FIXED(df, fade_ms=5000):

    pygame.mixer.init()

    paths = df['Path'].dropna().tolist()

    print("\n🔍 analyzing tracks...")
    meta = {p: _analyze_track(p) for p in tqdm(paths)}

    current = {"ch": None, "path": None, "start": None}
    running = {"on": True}
    lock = threading.Lock()


    # ---------- PREP ----------
    def prep(path, eq=(1,1,1)):
        y, sr = _trim_to_kick(path, meta[path]["start_sec"])
        y = _apply_eq(y, sr, *eq)
        return _to_sound(y)


    # ---------- PLAY ----------
    def play(path):

        snd = prep(path, (1,1,1))
        ch = snd.play()

        current["ch"] = ch
        current["path"] = path
        current["start"] = time.time()

        print(f"\n▶️ {path}")


    # ---------- TRANSITION ----------
    def transition():

        next_path = random.choice(paths)

        # 🔥 DO NOT KILL KICK
        snd_new = prep(next_path, (0.4, 0.7, 1.0))
        ch_new = snd_new.play()

        steps = 60
        dt = fade_ms / 1000 / steps

        for i in range(steps):

            t = i / steps

            fade_out = np.cos(t * np.pi/2)
            fade_in  = np.sin(t * np.pi/2)

            if current["ch"]:
                current["ch"].set_volume(fade_out)

            ch_new.set_volume(fade_in)

            time.sleep(dt)

        if current["ch"]:
            current["ch"].fadeout(300)

        current["ch"] = ch_new
        current["path"] = next_path
        current["start"] = time.time()

        print(f"\n⏭️ {next_path}")


    # ---------- AUTO LOOP ----------
    def loop():

        while running["on"]:

            time.sleep(0.5)

            if current["path"] is None:
                continue

            elapsed = time.time() - current["start"]
            dur = meta[current["path"]]["duration"]

            if elapsed >= dur - (fade_ms/1000 + 1):

                with lock:
                    transition()


    threading.Thread(target=loop, daemon=True).start()

    print("\n🎛 s=start | n=next | q=quit")

    while True:

        cmd = input("👉 ").strip().lower()

        if cmd == "s":
            with lock:
                play(random.choice(paths))

        elif cmd == "n":
            with lock:
                transition()

        elif cmd == "q":
            running["on"] = False
            if current["ch"]:
                current["ch"].fadeout(1000)
            break
```
