```python
# ---------- CONCAT VIDEOS BY CREATION TIME ----------
import subprocess
import os

def _vid_3001_i1_GET_concat_by_mov_time(
    folder_path,
    out_path="FULL_VIDEO.mov"
):
    files = [
        f for f in os.listdir(folder_path)
        if f.lower().endswith(".mov")
    ]

    records = []
    for f in files:
        full = os.path.join(folder_path, f)
        cmd = [
            "ffprobe", "-v", "error",
            "-show_entries", "format_tags=creation_time",
            "-of", "default=nw=1:nk=1",
            full
        ]
        result = subprocess.run(cmd, capture_output=True, text=True)
        ts = result.stdout.strip()
        records.append((ts, f))

    records.sort()

    concat_txt = os.path.join(folder_path, "_concat.txt")
    with open(concat_txt, "w") as fh:
        for _, f in records:
            fh.write(f"file '{os.path.join(folder_path, f)}'\n")

    subprocess.run([
        "ffmpeg", "-f", "concat", "-safe", "0",
        "-i", concat_txt,
        "-c", "copy",
        out_path
    ])
```
