```python
# -----######-----###### SMART GOPRO DJ AUDIO EXTRACTOR + CUTTER -----######-----######
from moviepy.editor import VideoFileClip
from pathlib import Path
from pydub import AudioSegment
import tempfile
import os

def _video_2907_djsetcombo_GET_fullpluscuts(folder_path):
    folder = Path(folder_path)
    video_files = sorted([f for f in folder.iterdir() if f.suffix.lower() == '.mp4'])

    if not video_files:
        print("❌ No .mp4 video files found.")
        return

    print("\n🔍 Proposed video progression:")
    for i, f in enumerate(video_files, 1):
        print(f"{i:02d}. {f.name}")

    confirm = input("\n❓ Is this the correct progression? (y/n): ").strip().lower()
    if confirm != 'y':
        print("⚠️ Please rename or reorder files and run again.")
        return

    print("\n🎧 Extracting MP3s per video...")
    temp_mp3s = []
    for video_file in video_files:
        print(f"🎬 Processing: {video_file.name}")
        try:
            # Extract audio to WAV (temp)
            clip = VideoFileClip(str(video_file))
            temp_wav_path = folder / f"{video_file.stem}_temp.wav"
            clip.audio.write_audiofile(str(temp_wav_path), codec='pcm_s16le')
            clip.close()

            # Convert WAV to MP3 cleanly
            clean_mp3_path = folder / f"{video_file.stem}_audio.mp3"
            audio_seg = AudioSegment.from_wav(temp_wav_path)
            audio_seg.export(clean_mp3_path, format="mp3")
            temp_mp3s.append(clean_mp3_path)

            # Clean up WAV
            os.remove(temp_wav_path)
        except Exception as e:
            print(f"❌ Failed to extract from {video_file.name}: {e}")
            continue

    if not temp_mp3s:
        print("⚠️ No valid MP3s extracted. Stopping.")
        return

    print("\n🔗 Stitching full MP3...")
    all_audio = sum([AudioSegment.from_mp3(p) for p in temp_mp3s])
    full_mp3_path = folder / "FULL_DJ_SET.mp3"
    all_audio.export(full_mp3_path, format="mp3")
    print(f"✅ FULL SET EXPORTED: {full_mp3_path}")

    # Delete temp MP3s
    for p in temp_mp3s:
        os.remove(p)

    # === INTERACTIVE CUTTER ===
    duration_ms = len(all_audio)
    total_sec = duration_ms / 1000
    hrs = int(total_sec // 3600)
    mins = int((total_sec % 3600) // 60)
    secs = int(total_sec % 60)

    print(f"\n🕓 FULL LENGTH: {hrs} hr {mins} min {secs} sec")

    while True:
        chop = input("\n✂️ Do you want to chop a time segment? (y/n): ").strip().lower()
        if chop != 'y':
            print("👋 Done.")
            break

        try:
            seg = input("⏱️ Enter range like `2.4 - 3.3` (hours): ").strip()
            start_hr, end_hr = map(float, seg.split("-"))
            start_ms = int(start_hr * 3600 * 1000)
            end_ms = int(end_hr * 3600 * 1000)
        except:
            print("⚠️ Invalid input. Try again.")
            continue

        if start_ms < 0 or end_ms > duration_ms or start_ms >= end_ms:
            print("❌ Invalid time range.")
            continue

        name = input("🎵 What should I name the segment (no .mp3)? ").strip()
        if not name:
            print("⚠️ Name is empty. Try again.")
            continue

        cut_path = folder / f"{name}.mp3"
        all_audio[start_ms:end_ms].export(cut_path, format="mp3")
        print(f"✅ Exported: {cut_path}")
```
