```python
# -----######-----###### FULL MP4 → MOV → MERGE + MP3 EXTRACTOR -----######-----###### #
from moviepy.editor import VideoFileClip, concatenate_videoclips
from pathlib import Path
from tqdm import tqdm
import sys

def _video_2507_mergecompressmp3_GET_final_video(folder_path):
    folder = Path(folder_path)

    # Step 1: Find .mp4 files (any case)
    video_files = sorted(
        [f for f in folder.iterdir() if f.is_file() and f.suffix.lower() == '.mp4'],
        key=lambda x: x.name.lower()
    )

    if not video_files:
        print("❌ No .mp4 files found in the folder.")
        return

    # Step 2: Show proposed order
    print("\n🧾 Proposed video order:")
    for i, f in enumerate(video_files, 1):
        print(f"{i}. {f.name}")
    sys.stdout.flush()
    input("\n❓ Is this the correct progression? (Press ENTER to continue or Ctrl+C to abort): ")

    # Step 3: Compress each .mp4 → .mov @ 50% size, 500k bitrate
    temp_movs = []
    for vid in tqdm(video_files, desc="🎬 Compressing MP4 to MOV"):
        clip = VideoFileClip(str(vid))
        new_size = (int(clip.w * 0.5), int(clip.h * 0.5))
        clip_resized = clip.resize(newsize=new_size)

        out_path = folder / f"{vid.stem}_25pct.mov"
        clip_resized.write_videofile(
            str(out_path),
            codec='libx264',
            audio_codec='aac',
            bitrate="500k",
            preset='medium',
            ffmpeg_params=["-movflags", "faststart"],
            threads=4,
            verbose=False,
            logger=None
        )
        temp_movs.append(out_path)
        clip.close()

    # Step 4: Merge all .mov clips
    print("\n📦 Merging MOV files...")
    clips = [VideoFileClip(str(mov)) for mov in temp_movs]
    final_video_path = folder / "final_merged_video.mov"
    final_clip = concatenate_videoclips(clips, method="compose")
    final_clip.write_videofile(str(final_video_path), codec='libx264', audio_codec='aac')
    final_clip.close()
    [c.close() for c in clips]

    # Step 5: Delete temp .mov
    print("🧹 Cleaning temporary MOVs...")
    for f in temp_movs:
        f.unlink(missing_ok=True)

    # Step 6: Extract audio as MP3
    print("🎧 Extracting audio from final video...")
    audio_path = final_video_path.with_suffix(".mp3")
    audio = VideoFileClip(str(final_video_path)).audio
    audio.write_audiofile(str(audio_path), bitrate="192k")
    audio.close()

    print(f"\n✅ Done.")
    print(f"🎞 Final video: {final_video_path.name}")
    print(f"🎧 Final MP3: {audio_path.name}")

    return final_video_path, audio_path
```
