```python
# -----######-----###### VIDEO INFO REPORT -----######-----######
import cv2
import os
from moviepy.editor import VideoFileClip
from pathlib import Path
from datetime import timedelta

def _video_0506_info_GET_attributes_report(video_path):
    video_path = Path(video_path)
    cap = cv2.VideoCapture(str(video_path))

    if not cap.isOpened():
        return f"❌ Cannot open video: {video_path}"

    width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
    height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
    fps = cap.get(cv2.CAP_PROP_FPS)
    frame_count = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
    duration_sec = frame_count / fps if fps > 0 else 0
    aspect_ratio = round(width / height, 2)

    cap.release()

    file_size_MB = round(os.path.getsize(video_path) / (1024 * 1024), 2)
    duration_hms = str(timedelta(seconds=int(duration_sec)))

    report = (
        f"📹 Video: {video_path.name}\n"
        f"⏱ Duration: {duration_hms} ({int(duration_sec)} seconds)\n"
        f"🖼 Resolution: {width}x{height}\n"
        f"📐 Aspect Ratio: {aspect_ratio}\n"
        f"🎞 FPS: {round(fps, 2)}\n"
        f"💾 File Size: {file_size_MB} MB"
    )
    return report


# -----######-----###### EXPORT IG STORY CLIP -----######-----######
def _video_0506_storyclip_GET_IGstory_mp4(video_path, duration=60, output_path=None):
    video_path = Path(video_path)
    clip = VideoFileClip(str(video_path))
    
    # Calculate middle point
    start_time = max(0, (clip.duration - duration) / 2)
    subclip = clip.subclip(start_time, start_time + duration)
    
    # Resize and crop to IG story format (9:16)
    subclip_resized = subclip.resize(height=1080).crop(x_center=subclip.w / 2, width=608)
    
    if output_path is None:
        output_path = video_path.with_name(video_path.stem + "_story.mp4")

    subclip_resized.write_videofile(str(output_path), codec="libx264", audio_codec="aac")
    return f"✅ Exported IG Story clip to: {output_path}"
```
