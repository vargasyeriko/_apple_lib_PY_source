```python
# -----######-----###### REDUCE VIDEO QUALITY TO 25% & EXPORT AS .MOV -----######-----###### #
from moviepy.editor import VideoFileClip
from pathlib import Path
from tqdm import tqdm

def video_2507_reducequality75_GET_mov(input_path, output_folder=None):
    """
    Reduces video quality to ~25% of the original and exports to MOV format.
    Keeps aspect ratio proportional.
    
    Parameters:
    - input_path: full path to .mp4 file
    - output_folder: folder to store .mov (default is same as input)
    
    Output:
    - Returns output_path (.mov)
    """

    input_path = Path(input_path)
    if not input_path.exists() or input_path.suffix.lower() != ".mp4":
        raise ValueError("Invalid .mp4 file path")

    output_folder = Path(output_folder) if output_folder else input_path.parent
    output_folder.mkdir(parents=True, exist_ok=True)

    clip = VideoFileClip(str(input_path))

    original_w, original_h = clip.size
    new_w, new_h = int(original_w * 0.5), int(original_h * 0.5)  # Resize to 50% resolution
    resized_clip = clip.resize(newsize=(new_w, new_h))

    output_path = output_folder / (input_path.stem + "_25pct.mov")

    print(f"\n🎥 Original resolution: {original_w}x{original_h}")
    print(f"📉 New resolution: {new_w}x{new_h}")
    print(f"💾 Saving to: {output_path.name}")

    # Estimate reduced bitrate (25% of estimated default)
    estimated_bitrate = int((clip.fps * new_w * new_h * 0.07) * 0.25)  # ~25% quality

    resized_clip.write_videofile(
        str(output_path),
        codec='libx264',
        bitrate=f"{estimated_bitrate}k",
        audio_codec='aac',
        threads=4,
        preset='medium',
        ffmpeg_params=["-movflags", "faststart"]
    )

    clip.close()
    return output_path
```
