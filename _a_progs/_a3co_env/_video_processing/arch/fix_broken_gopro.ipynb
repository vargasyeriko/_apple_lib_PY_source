```python
import subprocess
import os

def _video_2410_reencode_mov_GET_fixed(path_input):
    """
    Attempts to re-encode a corrupted MP4 to a new .mov format using ffmpeg.
    Will try to salvage as much as possible.
    """

    if not os.path.exists(path_input):
        print("❌ File does not exist.")
        return None

    path_out = os.path.splitext(path_input)[0] + "_REENCODED.mov"

    cmd = [
        "ffmpeg", "-y",
        "-err_detect", "ignore_err",  # skip bad frames
        "-fflags", "+genpts",
        "-i", path_input,
        "-c:v", "libx264",  # re-encode video to H.264
        "-c:a", "aac",      # re-encode audio
        "-movflags", "+faststart",
        path_out
    ]

    print(f"🚧 Re-encoding to MOV: {path_out}")
    result = subprocess.run(cmd, stdout=subprocess.PIPE, stderr=subprocess.PIPE)
    print(result.stderr.decode("utf-8")[:1000])
    
    if os.path.exists(path_out) and os.path.getsize(path_out) > 0:
        print(f"✅ Re-encoded .mov saved to: {path_out}")
        return path_out
    else:
        print("❌ Re-encoding failed.")
        return None
```

```python
# -----######-----###### FINAL EMERGENCY MOV RECOVERY -----######-----###### #
import subprocess
import os

def _video_2410_finalrescue_REENCODE_mov(path_input):
    """
    Emergency repair: Attempts to re-encode a severely broken GoPro .MP4 to .MOV
    Skips bad frames, rebuilds timeline and audio using ffmpeg.
    """
    if not os.path.exists(path_input):
        print("❌ File not found:", path_input)
        return None

    path_out = os.path.splitext(path_input)[0] + "_REENCODED_FINAL.mov"

    cmd = [
        "ffmpeg", "-y",
        "-fflags", "+genpts",  # guess timestamps
        "-err_detect", "ignore_err",  # skip corrupted frames
        "-analyzeduration", "100M",
        "-probesize", "100M",
        "-i", path_input,
        "-c:v", "libx264",  # re-encode video
        "-c:a", "aac",      # re-encode audio
        "-preset", "ultrafast",
        "-movflags", "+faststart",
        path_out
    ]

    print(f"🚧 Trying FINAL MOV rescue...\n→ Input: {path_input}\n→ Output: {path_out}")
    result = subprocess.run(cmd, stdout=subprocess.PIPE, stderr=subprocess.PIPE)
    print(result.stderr.decode("utf-8")[:1500])  # Show some output in case of issues

    if os.path.exists(path_out) and os.path.getsize(path_out) > 0:
        print(f"✅ Re-encoded .mov saved to:\n{path_out}")
        return path_out
    else:
        print("❌ Final re-encoding failed.")
        return None
```
