```python
# -----######-----###### MP4 INFO TABLE (CASE-INSENSITIVE) -----######-----######
import os
from pathlib import Path
import pandas as pd
from moviepy.editor import VideoFileClip

def _video_0506_info_GET_duration_size_table(input_folder):
    input_folder = Path(input_folder)
    video_data = []

    for file in input_folder.iterdir():
        if file.is_file() and file.suffix.lower() == ".mp4":
            try:
                clip = VideoFileClip(str(file))
                duration_sec = clip.duration
                size_gb = os.path.getsize(file) / (1024 ** 3)
                video_data.append({
                    "file_name": file.name,
                    "duration_sec": round(duration_sec, 2),
                    "duration_min": round(duration_sec / 60, 2),
                    "size_gb": round(size_gb, 3)
                })
                clip.reader.close()
                if clip.audio:
                    clip.audio.reader.close_proc()
            except Exception as e:
                video_data.append({
                    "file_name": file.name,
                    "duration_sec": "Error",
                    "duration_min": "Error",
                    "size_gb": "Error"
                })

    return pd.DataFrame(video_data)

#!#!#!#!#! RUNNING STATEMENTS #!#!#!#!#!
df_videos = _video_0506_info_GET_duration_size_table("/Users/yerik/Desktop/input_vids")
print(df_videos)
```
