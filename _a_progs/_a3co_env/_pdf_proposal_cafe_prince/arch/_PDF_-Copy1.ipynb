```python
def _pdf_0907_corecity_GET_clickablewords_pdf(output_path, content_lines, link_map):
    king_blue = HexColor("#3366CC")
    background_blue = HexColor("#EAF2FB")
    flier_paths = [flier_1, flier_2, flier_3, flier_4]

    # Professional serif font fallback
    try:
        pdfmetrics.registerFont(TTFont("Professional", "/usr/share/fonts/truetype/liberation/LiberationSerif-Regular.ttf"))
        font_name = "Professional"
    except:
        font_name = "Helvetica"

    c = canvas.Canvas(output_path, pagesize=A4)
    width, height = A4
    margin = 40

    # Frame + Background
    c.setStrokeColor(king_blue)
    c.setLineWidth(3)
    c.rect(margin, margin, width - 2 * margin, height - 2 * margin)
    c.setFillColor(background_blue)
    c.rect(margin + 2, margin + 2, width - 2 * margin - 4, height - 2 * margin - 4, fill=1, stroke=0)

    # Main body
    text = c.beginText(margin + 10, height - 60)
    text.setFont(font_name, 11)
    text.setFillColor(HexColor("#000000"))  # <-- FIXED HERE

    for line in content_lines:
        y = text.getY()

        for label, url in link_map.items():
            if label in line:
                x = margin + 10 + (line.index(label) * 5.9)
                c.linkURL(url, (x, y - 2, x + len(label) * 6, y + 10), relative=1, color=king_blue)
                break

        text.textLine(line)

        if text.getY() < 80:
            c.drawText(text)
            c.showPage()
            c.setFillColor(background_blue)
            c.rect(margin + 2, margin + 2, width - 2 * margin - 4, height - 2 * margin - 4, fill=1, stroke=0)
            text = c.beginText(margin + 10, height - 60)
            text.setFont(font_name, 11)
            text.setFillColor(HexColor("#000000"))  # <-- FIXED HERE

    c.drawText(text)

    # Add flier page (2x2 grid)
    c.showPage()
    flier_w = (width - 2 * margin - 20) / 2
    flier_h = (height - 2 * margin - 40) / 2
    positions = [
        (margin, height - margin - flier_h),
        (margin + flier_w + 20, height - margin - flier_h),
        (margin, margin + 20),
        (margin + flier_w + 20, margin + 20)
    ]

    for i, (path, pos) in enumerate(zip(flier_paths, positions), start=1):
        if os.path.exists(path):
            c.drawImage(ImageReader(path), pos[0], pos[1],
                        width=flier_w, height=flier_h,
                        preserveAspectRatio=True, anchor='c')
            c.setFont(font_name, 9)
            c.setFillColor(king_blue)
            c.drawString(pos[0] + 4, pos[1] - 12, f"flier {i}")

    c.save()
```

```python
def _pdf_0907_corecity_GET_clickablewords_pdf(output_path, content_lines, link_map):
    king_blue = HexColor("#3366CC")
    yellow = HexColor("#3366CC")
    green = HexColor("#3366CC")
    background_blue = HexColor("#EAF2FB")
    flier_paths = [flier_1, flier_2, flier_3, flier_4]

    try:
        pdfmetrics.registerFont(TTFont("Professional", "/usr/share/fonts/truetype/liberation/LiberationSerif-Regular.ttf"))
        font_name = "Professional"
    except:
        font_name = "Helvetica"

    c = canvas.Canvas(output_path, pagesize=A4)
    width, height = A4
    margin = 40
    page_count = 1

    def draw_framed_background(frame_color):
        c.setStrokeColor(frame_color)
        c.setLineWidth(3)
        c.rect(margin, margin, width - 2 * margin, height - 2 * margin)
        c.setFillColor(background_blue)
        c.rect(margin + 2, margin + 2, width - 2 * margin - 4, height - 2 * margin - 4, fill=1, stroke=0)

    draw_framed_background(king_blue)

    text = c.beginText(margin + 10, height - 60)
    text.setFont(font_name, 11)
    text.setFillColor(HexColor("#000000"))

    for line in content_lines:
        y = text.getY()

        for label, url in link_map.items():
            if label in line:
                x = margin + 10 + (line.index(label) * 5.9)
                c.linkURL(url, (x, y - 2, x + len(label) * 6, y + 10), relative=1, color=king_blue)
                break

        text.textLine(line)

        if text.getY() < 80:
            c.drawText(text)
            c.showPage()
            page_count += 1
            frame_color = yellow if page_count == 2 else green
            draw_framed_background(frame_color)
            text = c.beginText(margin + 10, height - 60)
            text.setFont(font_name, 11)
            text.setFillColor(HexColor("#000000"))

    c.drawText(text)

    # -------- Add Flier Grid Page (Blue Frame) -------- #
    c.showPage()
    draw_framed_background(king_blue)

    flier_w = (width - 2 * margin - 40) / 2
    flier_h = (height - 2 * margin - 60) / 2
    positions = [
        (margin + 5, height - margin - flier_h - 5),
        (margin + flier_w + 25, height - margin - flier_h - 5),
        (margin + 5, margin + 25),
        (margin + flier_w + 25, margin + 25)
    ]

    for i, (path, pos) in enumerate(zip(flier_paths, positions), start=1):
        if os.path.exists(path):
            c.drawImage(ImageReader(path), pos[0], pos[1],
                        width=flier_w, height=flier_h,
                        preserveAspectRatio=True, anchor='c')
            c.setFont(font_name, 9)
            c.setFillColor(king_blue)
            c.drawString(pos[0] + 4, pos[1] - 12, f"flier {i}")

    c.save()
    
_pdf_0907_corecity_GET_clickablewords_pdf(
    "Core_City_Community_Sundays_FINAL_with_Fliers.pdf",
    proposal_lines,
    gear_links
)
```
