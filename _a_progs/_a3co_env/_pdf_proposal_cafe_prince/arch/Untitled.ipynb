```python
# -----######-----######-----######-----######-----######-----
# 0_FNS — PDF Generator Function for Cafe Prince Party Proposal
# -----######-----######-----######-----######-----######-----

from reportlab.lib.pagesizes import A4
from reportlab.lib.units import inch
from reportlab.pdfgen import canvas
from reportlab.lib.colors import HexColor

def _pdf_0907_proposal_GET_cafeparty_blueframe(output_path):
    # Define king blue and light sky blue from logo
    king_blue = HexColor("#3366CC")
    background_blue = HexColor("#DCE6F9")

    # Setup canvas
    c = canvas.Canvas(output_path, pagesize=A4)
    width, height = A4

    # Draw blue border frame
    frame_margin = 30
    c.setStrokeColor(king_blue)
    c.setLineWidth(4)
    c.rect(frame_margin, frame_margin, width - 2*frame_margin, height - 2*frame_margin)

    # Fill background lightly
    c.setFillColor(background_blue)
    c.rect(frame_margin + 4, frame_margin + 4, width - 2*frame_margin - 8, height - 2*frame_margin - 8, fill=1, stroke=0)

    # Title
    c.setFillColor(king_blue)
    c.setFont("Helvetica-Bold", 26)
    c.drawCentredString(width / 2, height - 100, "Cafe Prince Party Proposal")

    # Subtitle
    c.setFont("Helvetica", 16)
    c.drawCentredString(width / 2, height - 130, "Core City — Pop-Up Music & Coffee Experience")

    # Body Text
    text = c.beginText(70, height - 180)
    text.setFont("Helvetica", 12)
    text.setFillColor(HexColor("#000000"))
    lines = [
        "We propose a unique Cafe Prince experience hosted in Core City.",
        "This community-forward event will blend café culture with a curated music lineup.",
        "",
        "Details:",
        "- Location: Core City, Detroit",
        "- Music: Live DJ sets by local artists",
        "- Vibe: Daytime lounge with evening energy",
        "- Amenities: Coffee cart, local pastries, pop-up vinyl crates",
        "",
        "Target Date: TBD between Sep–Oct 2025",
        "Audience: Young creative professionals, music lovers, neighbors",
        "",
        "This proposal outlines budget, talent, logistics, and marketing strategy."
    ]
    for line in lines:
        text.textLine(line)

    c.drawText(text)

    # Footer
    c.setFont("Helvetica-Oblique", 10)
    c.setFillColor(king_blue)
    c.drawCentredString(width / 2, 40, "Prepared by Yeriko – Custom Audio Integrations")

    c.save()
```

```python
def _pdf_0907_corecity_GET_clickablewords_pdf(output_path, content_lines, link_map):
    king_blue = HexColor("#3366CC")
    background_blue = HexColor("#DCE6F9")
    c = canvas.Canvas(output_path, pagesize=A4)
    width, height = A4
    margin = 30

    # Frame + background
    c.setStrokeColor(king_blue)
    c.setLineWidth(4)
    c.rect(margin, margin, width - 2 * margin, height - 2 * margin)
    c.setFillColor(background_blue)
    c.rect(margin + 2, margin + 2, width - 2 * margin - 4, height - 2 * margin - 4, fill=1, stroke=0)

    # Start text
    text = c.beginText(50, height - 60)
    text.setFont("Courier", 10)
    text.setFillColor("#000000")

    for line in content_lines:
        y = text.getY()

        for label, url in link_map.items():
            if label in line:
                x = 50 + (line.index(label) * 6)
                c.linkURL(url, (x, y - 2, x + len(label) * 6, y + 10), relative=1, color=king_blue)
                break

        text.textLine(line)

        # Page break
        if text.getY() < 60:
            c.drawText(text)
            c.showPage()
            c.setFillColor(background_blue)
            c.rect(margin + 2, margin + 2, width - 2 * margin - 4, height - 2 * margin - 4, fill=1, stroke=0)
            text = c.beginText(50, height - 60)
            text.setFont("Courier", 10)
            text.setFillColor("#000000")

    c.drawText(text)
    c.save()


_pdf_0907_corecity_GET_clickablewords_pdf("Core_City_Community_Sundays_FINAL.pdf", proposal_lines, gear_links)
```
