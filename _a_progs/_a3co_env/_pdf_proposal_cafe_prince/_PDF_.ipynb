```python
from reportlab.pdfgen import canvas
from reportlab.lib.pagesizes import A4
from reportlab.lib.colors import HexColor
from reportlab.lib.utils import ImageReader
from reportlab.pdfbase import pdfmetrics
from reportlab.pdfbase.ttfonts import TTFont
import os

def _pdf_0907_corecity_GET_clickablewords_pdf(output_path, content_lines, link_map):
    king_blue = HexColor("#3366CC")
    yellow = HexColor("#FFD700")
    green = HexColor("#66CC66")
    background_blue = HexColor("#EAF2FB")
    flier_paths = [flier_1, flier_2, flier_3, flier_4]

    # Font registration
    try:
        pdfmetrics.registerFont(TTFont("Professional", "/usr/share/fonts/truetype/liberation/LiberationSerif-Regular.ttf"))
        font_name = "Professional"
    except:
        font_name = "Helvetica"

    font_size = 11
    char_width = 5.9
    line_height = 14
    margin = 40
    width, height = A4
    c = canvas.Canvas(output_path, pagesize=A4)

    def draw_framed_background(color):
        c.setStrokeColor(color)
        c.setLineWidth(3)
        c.rect(margin, margin, width - 2 * margin, height - 2 * margin)
        c.setFillColor(background_blue)
        c.rect(margin + 2, margin + 2, width - 2 * margin - 4, height - 2 * margin - 4, fill=1, stroke=0)

    y = height - 60
    page = 1
    draw_framed_background(king_blue)

    for line in content_lines:
        c.setFont(font_name, font_size)
        c.setFillColor(HexColor("#000000"))
        c.drawString(margin + 10, y, line)

        # Overlay hyperlinks after drawing text
        for label, url in link_map.items():
            if label in line:
                x = margin + 10 + (line.index(label) * char_width)
                c.linkURL(url, (x, y - 2, x + len(label) * char_width, y + 10), relative=0)

        y -= line_height
        if y < 60:
            c.showPage()
            page += 1
            if page == 2:
                draw_framed_background(yellow)
            elif page == 3:
                draw_framed_background(green)
            else:
                draw_framed_background(king_blue)
            y = height - 60

    # Add flier page
    c.showPage()
    draw_framed_background(king_blue)
    flier_w = (width - 2 * margin - 40) / 2.1
    flier_h = (height - 2 * margin - 60) / 2.1
    positions = [
        (margin + 5, height - margin - flier_h - 5),
        (margin + flier_w + 25, height - margin - flier_h - 5),
        (margin + 5, margin + 25),
        (margin + flier_w + 25, margin + 25)
    ]

    for i, (path, pos) in enumerate(zip(flier_paths, positions), start=1):
        if os.path.exists(path):
            c.drawImage(ImageReader(path), pos[0], pos[1],
                        width=flier_w, height=flier_h,
                        preserveAspectRatio=True, anchor='c')
            c.setFont(font_name, 9)
            c.setFillColor(king_blue)
            c.drawString(pos[0] + 4, pos[1] - 12, f"flier {i}")

    c.save()
```

```python
_pdf_0907_corecity_GET_clickablewords_pdf(
    "Core_City_Community_Sundays_FINAL_with_Fliers.pdf",
    proposal_lines,
    gear_links
)
```
