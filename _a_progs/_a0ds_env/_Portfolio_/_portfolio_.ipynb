```python
from reportlab.pdfgen import canvas
from reportlab.lib.pagesizes import A4
from reportlab.lib.colors import HexColor
from reportlab.platypus import Paragraph
from reportlab.lib.styles import getSampleStyleSheet


# ---------------------------------------------------------
# PARAGRAPH WRAPPER
# ---------------------------------------------------------

def draw_paragraph(c, text, x, y, width):

    styles = getSampleStyleSheet()
    style = styles["Normal"]

    paragraph = Paragraph(text.strip(), style)

    w, h = paragraph.wrap(width, 800)

    paragraph.drawOn(c, x, y - h)

    return y - h


# ---------------------------------------------------------
# EXPERIENCE BLOCK
# ---------------------------------------------------------

def render_experience(c, y, title, desc, pdf_link, code_link, width):

    blue = HexColor("#1F4E79")

    c.setFont("Helvetica-Bold", 14)
    c.setFillColor(blue)
    c.drawString(60, y, title)

    y -= 18

    y = draw_paragraph(c, desc, 70, y, 450)

    y -= 18

    c.setFont("Helvetica-Oblique", 10)

    if pdf_link or code_link:

        if pdf_link:
            c.drawString(70, y, "Project")
            c.linkURL(pdf_link, (70, y-2, 120, y+10))

        if code_link:
            code_x = width - 120
            c.drawString(code_x, y, "Code")
            c.linkURL(code_link, (code_x, y-2, code_x+40, y+10))

        y -= 16

    y -= 20

    return y


# ---------------------------------------------------------
# BUILD PORTFOLIO
# ---------------------------------------------------------

def build_portfolio_pdf(output_path):

    width, height = A4
    c = canvas.Canvas(output_path, pagesize=A4)

    TOP_MARGIN = 80
    BOTTOM_MARGIN = 80

    y = height - TOP_MARGIN

    # HEADER
    c.setFont("Helvetica-Bold", 24)
    c.drawString(60, y, "Yeriko Vargas")

    y -= 25

    c.setFont("Helvetica", 13)
    c.drawString(60, y, "Data Scientist | Machine Learning Engineer | Sr. Data Analyst")

    y -= 10

    c.line(60, y, width-60, y)

    y -= 20

    # INTRO
    c.setFont("Helvetica", 11)

    y = draw_paragraph(c, intro_text, 60, y, 470)

    y -= 25

    # EXPERIENCE LOOP

    styles = getSampleStyleSheet()
    style = styles["Normal"]

    i = 1

    while True:

        desc = globals().get(f"exp_{i}_desc")

        if desc is None:
            break

        title = globals().get(f"exp_{i}_title", f"Project {i}")

        pdf_link = globals().get(f"exp_{i}_pdf_link")
        code_link = globals().get(f"exp_{i}_link_code")

        paragraph = Paragraph(desc.strip(), style)
        _, desc_height = paragraph.wrap(450, 800)

        estimated_height = 18 + desc_height + 18 + 16 + 20

        if y - estimated_height < BOTTOM_MARGIN:
            c.showPage()
            y = height - TOP_MARGIN

        y = render_experience(
            c,
            y,
            title,
            desc,
            pdf_link,
            code_link,
            width
        )

        i += 1

    c.save()

```
