```python
import pandas as pd

df = pd.DataFrame({
    "customer_id": [1, 2, 1, 3, 2, 1],
    "product": ["A", "B", "A", "C", "B", "C"],
    "price": [10, 20, 10, 30, 20, 30],
    "quantity": [1, 2, 3, 1, 1, 2]
})
```

```python
import duckdb

duckdb.query("""
    SELECT 
        customer_id,
        SUM(price * quantity) AS total_spent
    FROM df
    GROUP BY customer_id
    ORDER BY total_spent DESC
""").to_df()
```
