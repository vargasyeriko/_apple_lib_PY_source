```python
# ============================== 0_FNS ==============================

from pathlib import Path

from openpyxl import load_workbook
from tqdm.auto import tqdm


# -----######-----###### CORE IMPORTABLE FUNCTION -----######-----######
def _xlsx_1707_i1_GET_remove_exact_rows(
    xlsx_path: str,
    sheet_name: str,
    column_name: str,
    exact_value: str,
    case_sensitive: bool = False,
) -> int:
    """
    Physically delete Excel rows where a specified column exactly
    matches the supplied value, then overwrite the original XLSX file.
    """

    file_path = Path(xlsx_path).expanduser().resolve()

    if not file_path.exists():
        raise FileNotFoundError(f"File not found: {file_path}")

    workbook = load_workbook(file_path)

    if sheet_name not in workbook.sheetnames:
        raise KeyError(
            f"Sheet '{sheet_name}' not found. "
            f"Available sheets: {workbook.sheetnames}"
        )

    worksheet = workbook[sheet_name]

    headers = {
        str(cell.value).strip(): cell.column
        for cell in worksheet[1]
        if cell.value is not None
    }

    if column_name not in headers:
        raise KeyError(
            f"Column '{column_name}' not found. "
            f"Available columns: {list(headers)}"
        )

    column_number = headers[column_name]
    target = str(exact_value).strip()

    if not case_sensitive:
        target = target.casefold()

    rows_to_delete = []

    for row_number in tqdm(
        range(2, worksheet.max_row + 1),
        desc=f"Checking {column_name}",
        unit="row",
    ):
        value = worksheet.cell(
            row=row_number,
            column=column_number,
        ).value

        if value is None:
            continue

        cell_value = str(value).strip()

        if not case_sensitive:
            cell_value = cell_value.casefold()

        if cell_value == target:
            rows_to_delete.append(row_number)

    # Delete from bottom upward so row numbers do not shift.
    for row_number in tqdm(
        reversed(rows_to_delete),
        total=len(rows_to_delete),
        desc="Deleting Excel rows",
        unit="row",
    ):
        worksheet.delete_rows(
            row_number,
            amount=1,
        )

    workbook.save(file_path)
    workbook.close()

    print(f"Rows permanently removed: {len(rows_to_delete):,}")
    print(f"Workbook overwritten: {file_path}")

    return len(rows_to_delete)
```
