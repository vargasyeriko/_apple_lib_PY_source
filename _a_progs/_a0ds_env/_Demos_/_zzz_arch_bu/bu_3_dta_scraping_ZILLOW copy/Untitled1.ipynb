```python


# -----######-----######-----######-----######-----######
# _re_1204_i5_GET_df_rentals_FINAL
# FINAL RENTAL FETCH (CORRECT PARSER + ROBUST)
# -----######-----######-----######-----######-----######

import requests
import pandas as pd
from tqdm import tqdm
from datetime import datetime
import time


def _re_1204_i5_GET_df_rentals_FINAL(api_key):

    # ---------------- INPUT ----------------
    location = input("\n📍 Enter city (e.g. Detroit, MI): ").strip()
    n_rows = int(input("🔢 Number of listings: ").strip())

    url = "https://realtor16.p.rapidapi.com/search/forrent"

    headers = {
        "X-RapidAPI-Key": api_key.strip(),
        "X-RapidAPI-Host": "realtor16.p.rapidapi.com"
    }

    params = {
        "location": location,
        "limit": n_rows
    }

    # ---------------- REQUEST ----------------
    print("\n🌐 Fetching data...")
    r = requests.get(url, headers=headers, params=params)

    if r.status_code != 200:
        print(f"❌ Failed: {r.status_code}")
        print(r.text[:300])
        return pd.DataFrame()

    data = r.json()

    # ---------------- CORRECT EXTRACTION ----------------
    listings = data.get("properties", [])

    if not listings:
        print("⚠️ No listings found")
        return pd.DataFrame()

    # ---------------- BUILD DF ----------------
    rows = []

    for l in tqdm(listings, desc="🏗 Building DF"):

        try:
            rows.append({
                "address": l.get("location", {}).get("address", {}).get("line"),
                "price": l.get("list_price"),
                "beds": l.get("description", {}).get("beds"),
                "baths": l.get("description", {}).get("baths_consolidated"),
                "sqft": l.get("description", {}).get("sqft"),
                "type": l.get("description", {}).get("type"),
                "lat": l.get("location", {}).get("address", {}).get("coordinate", {}).get("lat"),
                "lon": l.get("location", {}).get("address", {}).get("coordinate", {}).get("lon"),
                "date_pulled": datetime.now()
            })
        except:
            continue

    df = pd.DataFrame(rows)

    # ---------------- SUMMARY ----------------
    print("\n📊 DONE")
    print(df.head())

    return df
```
