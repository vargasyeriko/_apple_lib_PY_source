```python
import numpy as np
import pandas as pd
from pathlib import Path


def generate_data(
        n_rows=50000,
        seed=42,
        save_csv=True,
        save_pkl=True,
        out_dir="data/raw",
        file_name="synthetic_dataset"
    ):
    """
    Generate random dataset suitable for ML models
    """

    np.random.seed(seed)

    # -----------------------------
    # FEATURES
    # -----------------------------

    age = np.random.randint(18, 70, n_rows)

    income = np.random.normal(65000, 20000, n_rows)
    income = np.clip(income, 20000, 200000)

    visits = np.random.poisson(5, n_rows)

    time_on_site = np.random.normal(8, 3, n_rows)
    time_on_site = np.clip(time_on_site, 1, 25)

    region = np.random.choice(
        ["north", "south", "east", "west"],
        n_rows
    )

    device = np.random.choice(
        ["mobile", "desktop", "tablet"],
        n_rows
    )

    # -----------------------------
    # TARGET VARIABLE
    # -----------------------------

    purchase_signal = (
        0.03 * visits +
        0.00002 * income +
        0.2 * (device == "desktop") +
        np.random.normal(0, 0.5, n_rows)
    )

    purchase = (purchase_signal > np.percentile(purchase_signal, 60)).astype(int)

    # -----------------------------
    # DATAFRAME
    # -----------------------------

    df = pd.DataFrame({
        "age": age,
        "income": income,
        "visits": visits,
        "time_on_site": time_on_site,
        "region": region,
        "device": device,
        "purchase": purchase
    })

    # -----------------------------
    # SAVE
    # -----------------------------

    out_dir = Path(out_dir)
    out_dir.mkdir(parents=True, exist_ok=True)

    if save_csv:
        df.to_csv(out_dir / f"{file_name}.csv", index=False)

    if save_pkl:
        df.to_pickle(out_dir / f"{file_name}.pkl")

    print(f"\nGenerated dataset with {n_rows} rows")

    return df
```

```python
df = generate_data(
    n_rows=50000
)
```
