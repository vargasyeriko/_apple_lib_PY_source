```python
import numpy as np
import pandas as pd

np.random.seed(42)

# -----------------------------
# SIMULATE DATA
# -----------------------------
n = 5000

df = pd.DataFrame({
    'weight': np.random.uniform(1, 10, n),
    'rocket_power': np.random.uniform(1, 10, n),
    'aerodynamic': np.random.uniform(1, 10, n)
})

# TRUE RELATION (hidden)
noise = np.random.normal(0, 1, n)

df['Y'] = (
    20.5 * df['weight'] +
    40.0 * df['rocket_power'] -
    1.5 * df['aerodynamic'] +
    3 +
    noise
)

df.head()

# WE are creating the y true and explanatory 
```

```python
from sklearn.linear_model import LinearRegression
```

```python
x = 4 
x = 1+1+1+1 # : X = matrix 
```

```python
# -----------------------------
# FEATURES + TARGET -> MY MODEL is 
# -----------------------------
X = df[['weight', 'rocket_power', 'aerodynamic']]
y = df['Y']

#y = x_1 + x_2 + x_3 =>>>>.   y = (x*c) +e 

# Y = x + e (error)
model = LinearRegression()
model.fit(X, y)
```

```python
# -----------------------------
# PREDICTIONS
# -----------------------------
df['Y_hat'] = model.predict(X)
#df['error'] = df['Y_hat'] - df['Y']
```

```python
df['err'] = df['Y'] - df['Y_hat'] 
```

```python
# -----------------------------
# OUTPUT
# -----------------------------
print("COEFFICIENTS:")
print("weight:", model.coef_[0])
print("rocket_power:", model.coef_[1])
print("aerodynamic:", model.coef_[2])
print("intercept:", model.intercept_)

print("\nHEAD:")
print(df[['Y','Y_hat','error']].head())
```
