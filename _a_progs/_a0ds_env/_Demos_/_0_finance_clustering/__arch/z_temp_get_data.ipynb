```python
# -----######-----###### CORE FUNCTION -----######-----######

import yfinance as yf
import pandas as pd

def _yf_0704_i2_GET_df_prices_cols(
    tickers,
    start,
    end,
    interval="1d",
    cols=None   # ["Open","High","Low","Close","Adj Close","Volume"]
):
    
    df = yf.download(
        tickers=tickers,
        start=start,
        end=end,
        interval=interval,
        group_by="ticker",
        auto_adjust=False,
        prepost=False,
        threads=True
    )
    
    # ---- handle single ticker vs multi ticker
    if isinstance(df.columns, pd.MultiIndex):
        
        if cols is not None:
            df = df.loc[:, (slice(None), cols)]
            
        # flatten columns → cleaner for ML workflows
        df.columns = [
            f"{tic}_{col}" for tic, col in df.columns
        ]
    
    else:
        if cols is not None:
            df = df[cols]
    
    print(f"\nRows: {len(df)} | Interval: {interval}")
    print(f"Columns: {list(df.columns)[:10]} ...")
    
    return df
```

```python
tickers = [
    "AAPL","MSFT","NVDA","AMZN","GOOGL",
    "TSLA","META","JPM","V","UNH"
]

# ---- ONLY CLOSE + VOLUME
df = _yf_0704_i2_GET_df_prices_cols(
    tickers,
    start="2024-01-01",
    end="2025-01-01",
    interval="1d",
    cols=["Close", "Volume"]
)

df.head()
```
