# IBKR API — Pull Market Data to Your Laptop

### Prerequisites
1. **TWS** (Trader Workstation) or **IB Gateway** must be running and logged in
2. API connections enabled: TWS → File → Global Configuration → API → Settings → **Enable ActiveX and Socket Clients** ✓
3. Socket port = **7497** (paper) or **7496** (live)
4. Install the library: `pip install ib_insync`

---

```python
from ib_insync import *
```

```python
# pip install ib_insync      ← run this once if you haven't
from ib_insync import *
import pandas as pd

# ── Connect to TWS / IB Gateway ──────────────────────────
ib = IB()
ib.connect("127.0.0.1",      # localhost (your machine)
           7497,              # 7497 = paper trading, 7496 = live
           clientId=1)

print(f"Connected: {ib.isConnected()}")
```

## Fetch Historical Daily Bars

```python
# ── Define what we want ────────────────────────────────
contract = Stock("AAPL", "SMART", "USD")

# ── Request 1 year of daily bars ──────────────────────
bars = ib.reqHistoricalData(
    contract,
    endDateTime="",           # "" = up to right now
    durationStr="1 Y",        # how far back: 1 year
    barSizeSetting="1 day",   # each bar = 1 day
    whatToShow="TRADES",      # price type
    useRTH=True,              # regular trading hours only
    formatDate=1,
)

# ── Convert to a pandas DataFrame ─────────────────────
df = util.df(bars)
df.set_index("date", inplace=True)

print(f"Got {len(df)} daily bars for AAPL")
df.tail()
```

## Quick Plot

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 4))
plt.plot(df.index, df["close"], color="#3B82F6", linewidth=1)
plt.title("AAPL — Daily Close (via IBKR API)", fontsize=13, fontweight="bold")
plt.ylabel("Price (USD)")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## Disconnect

```python
ib.disconnect()
print("Disconnected.")
```

### Quick Reference

| Parameter | Options |
|-----------|---------|
| **Port** | `7497` paper · `7496` live |
| **durationStr** | `"30 D"` · `"6 M"` · `"1 Y"` · `"2 Y"` |
| **barSizeSetting** | `"1 min"` · `"5 mins"` · `"1 hour"` · `"1 day"` · `"1 week"` |
| **whatToShow** | `"TRADES"` · `"MIDPOINT"` · `"BID"` · `"ASK"` |
| **Contract types** | `Stock("AAPL","SMART","USD")` · `Forex("EURUSD")` · `Future("ES","CME")` |
