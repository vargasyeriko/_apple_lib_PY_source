```python
# -----######-----######-----######-----######-----######
# _re_1204_i2_GET_df_deal_engine_clean
# DEAL ENGINE (CLEAN FEATURES + ORDERED OUTPUT)
# -----######-----######-----######-----######-----######

import pandas as pd
import numpy as np
from tqdm import tqdm


def _re_1204_i2_GET_df_deal_engine_clean(df):

    df = df.copy()
    tqdm.pandas()

    # -----------------------------
    # 1. CLEAN TYPES
    # -----------------------------
    df['list_price'] = pd.to_numeric(df['list_price'], errors='coerce')
    df['description_sqft'] = pd.to_numeric(df['description_sqft'], errors='coerce')
    df['description_lot_sqft'] = pd.to_numeric(df['description_lot_sqft'], errors='coerce')
    df['description_beds'] = pd.to_numeric(df['description_beds'], errors='coerce')
    df['description_baths_consolidated'] = pd.to_numeric(df['description_baths_consolidated'], errors='coerce')
    df['price_reduced_amount'] = pd.to_numeric(df['price_reduced_amount'], errors='coerce')

    df['list_date'] = (
        pd.to_datetime(df['list_date'], errors='coerce')
        .dt.tz_convert(None)
    )

    # -----------------------------
    # 2. DERIVED FEATURES
    # -----------------------------
    today = pd.Timestamp.today()

    df['price_per_sqft'] = df['list_price'] / df['description_sqft']
    df['discount_pct'] = df['price_reduced_amount'] / df['list_price']
    df['days_on_market'] = (today - df['list_date']).dt.days

    df['discount_pct'] = df['discount_pct'].fillna(0)
    df['days_on_market'] = df['days_on_market'].fillna(0)

    # -----------------------------
    # 3. NORMALIZATION
    # -----------------------------
    def _norm(series):
        return (series - series.min()) / (series.max() - series.min() + 1e-9)

    # -----------------------------
    # 4. FEATURE SCORES
    # -----------------------------
    structure = (
        _norm(df['description_sqft']) * 0.5 +
        (1 - _norm(df['price_per_sqft'])) * 0.5
    )

    dynamics = (
        _norm(df['discount_pct']) * 0.7 +
        (1 - _norm(df['list_price'])) * 0.3
    )

    context = np.ones(len(df)) * 0.5

    friction = (
        _norm(df['days_on_market']) * 0.7 +
        df['flags_is_foreclosure'].fillna(False).astype(int) * 0.3
    )

    optionality = (
        _norm(df['description_lot_sqft']) * 0.5 +
        _norm(df['description_sqft']) * 0.5
    )

    # -----------------------------
    # 5. FINAL SCORES
    # -----------------------------
    df['score_structure'] = structure * 100
    df['score_dynamics'] = dynamics * 100
    df['score_context'] = context * 100
    df['score_friction'] = friction * 100
    df['score_optionality'] = optionality * 100

    df['deal_score'] = (
        0.25 * df['score_structure'] +
        0.30 * df['score_dynamics'] +
        0.20 * df['score_context'] +
        0.15 * (100 - df['score_friction']) +
        0.10 * df['score_optionality']
    )

    # -----------------------------
    # 6. SELECT + ORDER COLUMNS (🔥 KEY FIX)
    # -----------------------------
    cols_final = [

        # ----- IDENTIFIERS -----
        'listing_id',
        'location_address_city',
        'location_address_postal_code',
        'location_address_line',

        # ----- STRUCTURE -----
        'list_price',
        'description_sqft',
        'description_lot_sqft',
        'description_beds',
        'description_baths_consolidated',
        'price_per_sqft',

        # ----- DYNAMICS -----
        'price_reduced_amount',
        'discount_pct',

        # ----- FRICTION -----
        'list_date',
        'days_on_market',
        'flags_is_foreclosure',

        # ----- SCORES -----
        'score_structure',
        'score_dynamics',
        'score_context',
        'score_friction',
        'score_optionality',
        'deal_score'
    ]

    # keep only existing cols (safe)
    cols_final = [c for c in cols_final if c in df.columns]

    df = df[cols_final].copy()

    # -----------------------------
    # 7. SORT
    # -----------------------------
    df = df.sort_values('deal_score', ascending=False).reset_index(drop=True)

    return df
```
