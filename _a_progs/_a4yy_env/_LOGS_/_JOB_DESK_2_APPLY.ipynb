```python
# -----######-----###### CORE TABLE FUNCTION -----######-----######

import pandas as pd

def _resume_2503_table_GET_df():

    data = [
        ["Wiingy", "Dec 2023", "Present", None, "Contract", "Remote", "Current role"],
        ["Ford", "Mar 2022", "Nov 2023", None, "Contract", "Dearborn, MI", "Contract completed"],
        ["Soothsayer Analytics", "May 2021", "Feb 2022", None, "Full-time", "Livonia, MI", "Career progression"],
        ["Official Driving School", "Aug 2020", "Apr 2021", None, "Full-time", "Royal Oak, USA", "Transition to data roles"],
        ["Fiat Chrysler Automotive", "Jun 2019", "Apr 2020", None, "Contract", "Auburn Hills, MI", "Contract completed"],
        ["Beaumont Hospital (Biostatistics)", "Jul 2018", "Sep 2018", None, "Internship", "Royal Oak, MI", "Internship completed"]
    ]

    cols = [
        "Company",
        "Start Date",
        "End Date",
        "Duration (Months)",
        "Employment Type",
        "Location",
        "Reason for Leaving"
    ]

    df = pd.DataFrame(data, columns=cols)

    # ---- Duration calculation ----
    def _calc_months(row):
        if row["End Date"] == "Present":
            return pd.Timestamp.today().to_period('M') - pd.Period(row["Start Date"], freq='M')
        else:
            return pd.Period(row["End Date"], freq='M') - pd.Period(row["Start Date"], freq='M')

    df["Duration (Months)"] = df.apply(lambda r: _calc_months(r).n if pd.notna(_calc_months(r)) else None, axis=1)

    # ---- Totals ----
    total_months = df["Duration (Months)"].sum()
    total_years = round(total_months / 12, 2)

    df_total = pd.DataFrame([
        ["TOTAL", "", "", total_months, "", "", ""],
        ["TOTAL YEARS", "", "", total_years, "", "", ""]
    ], columns=cols)

    df_final = pd.concat([df, df_total], ignore_index=True)

    return df_final

#!#!#!#!#! RUNNING STATEMENTS #!#!#!#!#!

df_resume = _resume_2503_table_GET_df()
df_resume
```
