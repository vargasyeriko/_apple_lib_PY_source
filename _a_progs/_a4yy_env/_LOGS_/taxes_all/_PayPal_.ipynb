```python
import pandas as pd

df = pd.read_csv('_2025_PayPal_.csv')
```

# COMPLETED transactions money coming in 

```python
# remove commas BEFORE converting
df['Amount'] = df['Amount'].astype(str).str.replace(',', '')

# convert to numeric
df['Amount'] = pd.to_numeric(df['Amount'], errors='coerce')

# filter completed
df_completed = df[df['Status'] == 'Completed'].copy()

# money in
df_in = df_completed[df_completed['Amount'] > 0].copy()
```

```python
# sort biggest first
df_in = df_in.sort_values(by='Amount', ascending=False).reset_index(drop=True)

# total
total_in = df_in['Amount'].sum()

# remove dad transactions
df_in_clean = df_in[
    ~df_in['Name'].str.contains('Erich Luckerbauer', case=False, na=False)
].copy()

# recompute total
total_in_clean = df_in_clean['Amount'].sum()

print("TOTAL INCOME (CLEAN):", total_in_clean)
# df_in_clean
```

# DEDUCTIONS 

```python
# money out (expenses)
df_out = df_completed[df_completed['Amount'] < 0].copy()
```

```python
df_out['Amount_abs'] = df_out['Amount'].abs()
```

```python
total_out = df_out['Amount'].sum()
print("TOTAL EXPENSES:", total_out)
print("TOTAL EXPENSES (ABS):", df_out['Amount'].abs().sum())
```
