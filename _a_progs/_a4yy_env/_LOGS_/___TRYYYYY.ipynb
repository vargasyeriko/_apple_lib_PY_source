```python
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle

# ---------------------------------
# CREATE FIGURE
# ---------------------------------
fig, ax = plt.subplots(figsize=(6, 6))

# ---------------------------------
# DEFINE GRID (X = bandwidth, Y = flatness)
# ---------------------------------

# X divisions (4 bandwidth zones)
x_bins = [0, 1, 2, 3, 4]

# Y divisions (2 flatness zones)
y_bins = [0, 1, 2]

# ---------------------------------
# COLOR MATRIX (YOUR LOGIC)
# bottom row = tonal
# top row = noisy
# ---------------------------------

colors = [
    ['#ADD8FF', '#3EA4FF', '#005EFF', '#113C70'],  # tonal
    ['#FFE8CF', '#FFC78C', '#FFA64F', '#FF7600']   # noisy
]

labels = [
    ['Sub', 'Groove Low', 'Mix Body', 'Hi-End'],
    ['Soft Noise', 'Noisy Groove', 'Distortion', 'Harsh']
]

# ---------------------------------
# DRAW GRID
# ---------------------------------

for y in range(2):
    for x in range(4):
        rect = Rectangle((x, y), 1, 1,
                         facecolor=colors[y][x],
                         edgecolor='black')
        ax.add_patch(rect)

        ax.text(x + 0.5, y + 0.5,
                labels[y][x],
                ha='center', va='center',
                fontsize=9, color='black')

# ---------------------------------
# AXES FORMATTING
# ---------------------------------

ax.set_xlim(0, 4)
ax.set_ylim(0, 2)

ax.set_xticks([0.5, 1.5, 2.5, 3.5])
ax.set_xticklabels([
    '<2000', '2000–3000', '3000–4200', '4200–5200+'
])

ax.set_yticks([0.5, 1.5])
ax.set_yticklabels([
    'Tonal (Flatness < 0.09)',
    'Noisy (Flatness ≥ 0.09)'
])

ax.set_xlabel('Spectral Bandwidth (Shape → Spread)')
ax.set_ylabel('Spectral Flatness (Texture → Noise)')

ax.set_title('Spectral Shape–Texture Decision Map')

plt.grid(False)
plt.tight_layout()
plt.show()
```
