```python
import pandas as pd
import matplotlib.pyplot as plt

# Example data
data = {
    'Date': pd.date_range('2025-08-11', periods=5),
    'AAPL': [210, 213, 215, 217, 220],
    'MSFT': [420, 425, 430, 435, 440],
    'TSLA': [700, 720, 740, 750, 770]
}
df = pd.DataFrame(data)

# Normalize to first day
df_norm = df.set_index('Date').apply(lambda x: (x / x.iloc[0]) * 100)

# Plot
plt.figure(figsize=(8,5))
for col in df_norm.columns:
    plt.plot(df_norm.index, df_norm[col], label=col)

plt.title("Stock Growth (Normalized to Day 1 = 100)")
plt.ylabel("Growth Index")
plt.legend()
plt.grid(True)
plt.show()
```
