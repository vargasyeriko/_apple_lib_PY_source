```python
import pandas as pd
import sqlite3
import re

conn = sqlite3.connect("/Users/yerik/Library/Application Support/AddressBook/AddressBook-v22.abcddb")

df = pd.read_sql("""
SELECT 
    c.ZFIRSTNAME,
    c.ZLASTNAME,
    p.ZFULLNUMBER
FROM ZABCDRECORD c
LEFT JOIN ZABCDPHONENUMBER p ON c.Z_PK = p.ZOWNER
""", conn)

conn.close()

df["phone_clean"] = df["ZFULLNUMBER"].apply(
    lambda x: re.sub(r"\D", "", str(x)) if pd.notnull(x) else None
)

df["name"] = (df["ZFIRSTNAME"].fillna("") + " " + df["ZLASTNAME"].fillna("")).str.strip()

df = df[["name", "phone_clean"]]

df.head(50)
```
